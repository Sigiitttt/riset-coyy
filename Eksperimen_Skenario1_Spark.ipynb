```python
# -*- coding: utf-8 -*-
"""skin_lesion_EfficientNetB0_MultiSeed_Stability.py

Automatically structured for Google Colab.

# Evaluating Class-Wise Performance Stability of EfficientNet-B0 for
# Skin Lesion Classification Across Controlled Random Seeds

Notebook ini adalah versi revisi dari pipeline EfficientNet-B0 sebelumnya
(`Jurnal2.py`). Fokus penelitian BUKAN lagi membandingkan pipeline Basic vs
Spark dan BUKAN membuktikan bahwa Spark meningkatkan performa. Fokus
penelitian ini adalah **mengukur stabilitas performa EfficientNet-B0
terhadap stochastic variation** menggunakan lima controlled random seed
yang telah ditentukan (42, 123, 2026, 3407, 8901).

Dataset, arsitektur EfficientNet-B0, preprocessing utama, class definition,
serta prosedur leakage detection & deduplication dipertahankan SAMA PERSIS
dari pipeline asli. Train/validation/test split TIDAK berubah dan TIDAK
dipengaruhi oleh seed -- seed hanya memengaruhi proses stokastik training
(inisialisasi classifier head, augmentasi, weighted sampling, urutan
mini-batch, dropout).

Interpretasi hasil penelitian ini adalah:
    "controlled random seeds are used to quantify stochastic performance
    variability and class-wise stability."
BUKAN:
    "multi-seed improves model performance."

Tidak ada seed yang dipilih sebagai "hasil terbaik" -- hasil utama adalah
statistik agregat (mean +/- SD, CV, 95% CI) dari kelima seed.

## 1. Install Library
"""

!pip install -q timm imagehash pyspark grad-cam captum lime shap
print('Semua library terinstall')

"""## 2. Import Library"""

import os
import sys
import platform
import random
import warnings
import logging

# Bungkam peringatan HuggingFace Hub dan symlink di Kaggle
os.environ['HF_HUB_DISABLE_SYMLINKS_WARNING'] = '1'
logging.getLogger('huggingface_hub').setLevel(logging.ERROR)

# Pastikan JAVA_HOME terdeteksi dengan benar untuk PySpark di Kaggle
if 'JAVA_HOME' not in os.environ:
    for jpath in ['/usr/lib/jvm/default-java', '/usr/lib/jvm/java-11-openjdk-amd64', '/usr/lib/jvm/java-17-openjdk-amd64']:
        if os.path.exists(jpath):
            os.environ['JAVA_HOME'] = jpath
            break
import time
import math
from datetime import datetime
from getpass import getpass
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd

import cv2
from PIL import Image

import hashlib
import imagehash

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
import torchvision
import torchvision.transforms as transforms
import timm

from sklearn.metrics import (
    roc_auc_score, roc_curve, auc, accuracy_score,
    precision_recall_fscore_support, confusion_matrix, classification_report
)
from sklearn.preprocessing import label_binarize
from scipy import stats as sp_stats

import matplotlib.pyplot as plt
import seaborn as sns

# ======================================================================
# [MULTI-SEED EXPERIMENT] Determinism setup that must happen BEFORE any
# CUDA context is created. CUBLAS_WORKSPACE_CONFIG is required by PyTorch
# for deterministic CUDA matmul/conv algorithms when
# torch.use_deterministic_algorithms(True) is requested.
# ======================================================================
os.environ.setdefault('CUBLAS_WORKSPACE_CONFIG', ':4096:8')

"""## 3. Aktivasi GPU"""

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device : {DEVICE}')
if DEVICE.type == 'cuda':
    print(f'GPU    : {torch.cuda.get_device_name(0)}')
    print(f'VRAM   : {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB')
else:
    print('GPU tidak aktif! Runtime -> Change Runtime Type -> GPU')

"""## 4. Load Dataset (Kaggle Native)

Di Kaggle Notebook, dataset TIDAK perlu diunduh manual melalui API token.
Cukup tambahkan 3 dataset berikut via menu "+ Add Input" di panel kanan atas:
1. wanderdust/skin-lesion-analysis-toward-melanoma-detection
2. kmader/skin-cancer-mnist-ham10000
3. muhammadaliviansidiq/isic-msk-1-2

Fungsi di bawah secara cerdas dan otomatis mendeteksi lokasi folder dan file di `/kaggle/input/`
tanpa khawatir perbedaan struktur subfolder!
"""

def resolve_path(expected_path, fallback_search_name=None):
    if os.path.exists(expected_path):
        return expected_path
    if fallback_search_name:
        for root in ['/kaggle/input', '/content/datasets', '.']:
            if os.path.exists(root):
                for dirpath, dirnames, filenames in os.walk(root):
                    if fallback_search_name in dirnames or fallback_search_name in filenames:
                        return os.path.join(dirpath, fallback_search_name)
    return expected_path

# Base search roots di Kaggle
KAG_INPUT = '/kaggle/input'

# Cari file HAM10000 metadata
MET_CSV = resolve_path(
    f'{KAG_INPUT}/skin-cancer-mnist-ham10000/HAM10000_metadata.csv',
    'HAM10000_metadata.csv'
)
BASE_DIR = os.path.dirname(MET_CSV) if (MET_CSV and os.path.exists(MET_CSV)) else f'{KAG_INPUT}/skin-cancer-mnist-ham10000'
print(f"HAM10000 Metadata terdeteksi di: {MET_CSV}")

# ============================================================
# SKENARIO IDENTIFICATION & BENCHMARK SETUP
# ============================================================
SCENARIO_ID             = 'scenario_1_spark_dataprep'
SCENARIO_LABEL          = 'Skenario 1 (PySpark UDF Data Management + PyTorch DL)'
DATAPREP_ENGINE_LABEL   = 'PySpark UDF & DataFrame (Distributed JVM)'
INFERENCE_ENGINE_LABEL  = 'PyTorch DataLoader (Batch 32) + Scikit-Learn'

OUTPUT_DIR = f'/kaggle/working/output_{SCENARIO_ID}'
os.makedirs(OUTPUT_DIR, exist_ok=True)
DRIVE_OUTPUT_DIR = None
print(f"Direktori Output Kaggle: {OUTPUT_DIR}")

t_pipeline_start = time.time()

NUM_CLASSES = 3
CLASS_NAMES = ['nevus', 'melanoma', 'seborrheic_keratosis']

# ------------------------------------------------------------------
# DATA_PREP_SEED is a FIXED constant used ONLY to order/shuffle the raw
# dataframe before leakage/dedup cleanup. It is intentionally decoupled
# from the SEEDS list used for the training experiment below, so that
# editing the training seed list can never accidentally change dataset
# composition or row order. Must stay 42 to reproduce the same train_df
# row order as the original baseline pipeline (Jurnal2.py).
# ------------------------------------------------------------------
DATA_PREP_SEED = 42

# Resolusi direktori ISIC 2017 & HAM10000 dengan fail-safe search
ISIC_TRAIN_DIR = resolve_path(f'{KAG_INPUT}/skin-lesion-analysis-toward-melanoma-detection/skin-lesions/train', 'train')
ISIC_VAL_DIR   = resolve_path(f'{KAG_INPUT}/skin-lesion-analysis-toward-melanoma-detection/skin-lesions/valid', 'valid')
ISIC_TEST_DIR  = resolve_path(f'{KAG_INPUT}/skin-lesion-analysis-toward-melanoma-detection/skin-lesions/test', 'test')
ISIC_MSK_DIR   = resolve_path(f'{KAG_INPUT}/isic-msk-1-2', 'isic-msk-1-2')
HAM_META_CSV   = MET_CSV

ham_part1 = resolve_path(f'{KAG_INPUT}/skin-cancer-mnist-ham10000/HAM10000_images_part_1', 'HAM10000_images_part_1')
ham_part2 = resolve_path(f'{KAG_INPUT}/skin-cancer-mnist-ham10000/HAM10000_images_part_2', 'HAM10000_images_part_2')
HAM_IMAGE_DIRS = [d for d in [ham_part1, ham_part2] if d and os.path.exists(d)]

print(f"ISIC Train Dir: {ISIC_TRAIN_DIR} (exists: {os.path.exists(ISIC_TRAIN_DIR) if ISIC_TRAIN_DIR else False})")
print(f"ISIC Val Dir  : {ISIC_VAL_DIR} (exists: {os.path.exists(ISIC_VAL_DIR) if ISIC_VAL_DIR else False})")
print(f"ISIC Test Dir : {ISIC_TEST_DIR} (exists: {os.path.exists(ISIC_TEST_DIR) if ISIC_TEST_DIR else False})")
print(f"ISIC MSK Dir  : {ISIC_MSK_DIR} (exists: {os.path.exists(ISIC_MSK_DIR) if ISIC_MSK_DIR else False})")
print(f"HAM Image Dirs: {HAM_IMAGE_DIRS}")

ISIC_FOLDER_MAP = {
    'nevus':                {'dx': 'nv',  'label': 0},
    'melanoma':             {'dx': 'mel', 'label': 1},
    'seborrheic_keratosis': {'dx': 'bkl', 'label': 2}
}

def parse_isic_split(base_dir, split_name):
    data = []
    for folder_name, info in ISIC_FOLDER_MAP.items():
        folder_path = os.path.join(base_dir, folder_name)
        if os.path.exists(folder_path):
            for img_name in os.listdir(folder_path):
                if img_name.lower().endswith(('.jpg', '.png', '.jpeg')):
                    data.append({
                        'image_path': os.path.join(folder_path, img_name),
                        'dx':         info['dx'],
                        'label':      info['label'],
                        'source':     f'ISIC2017_{split_name}'
                    })
        else:
            print(f'  Folder tidak ditemukan: {folder_path}')
    return pd.DataFrame(data)

df_isic_train = parse_isic_split(ISIC_TRAIN_DIR, 'Train')
df_msk_train  = parse_isic_split(ISIC_MSK_DIR,   'MSK')
df_isic_val   = parse_isic_split(ISIC_VAL_DIR,   'Val')
df_isic_test  = parse_isic_split(ISIC_TEST_DIR,  'Test')

df_ham_full = pd.read_csv(HAM_META_CSV)
df_ham = df_ham_full[df_ham_full['dx'].isin(['mel', 'bkl'])].copy()

ham_path_dict = {}
for d in HAM_IMAGE_DIRS:
    if os.path.exists(d):
        for filename in os.listdir(d):
            if filename.endswith('.jpg'):
                ham_path_dict[filename.split('.')[0]] = os.path.join(d, filename)

df_ham['image_path'] = df_ham['image_id'].map(ham_path_dict)
df_ham = df_ham.dropna(subset=['image_path'])
df_ham['label']  = df_ham['dx'].map({'mel': 1, 'bkl': 2})
df_ham['source'] = 'HAM10000_Extra'
df_ham = df_ham[['image_path', 'dx', 'label', 'source']]

val_df   = df_isic_val.copy()
test_df  = df_isic_test.copy()
train_df = pd.concat([df_isic_train, df_msk_train, df_ham], ignore_index=True)
train_df = train_df.sample(frac=1, random_state=DATA_PREP_SEED).reset_index(drop=True)

print('--- RINGKASAN DATASET ---')
print(f'Train : {len(train_df):5,} samples')
print(f'Val   : {len(val_df):5,} samples')
print(f'Test  : {len(test_df):5,} samples')

"""## 4b. Pemeriksaan Data Leakage (Duplicate & Near-Duplicate Detection)

Bagian ini TIDAK DIUBAH dari pipeline asli (sudah benar dan wajib
dipertahankan sesuai instruksi). Exact duplicate dideteksi lewat file hash
MD5, near-duplicate dideteksi lewat perceptual hash (pHash) + jarak Hamming.
Leakage lintas split (train vs val/test) dibersihkan dari sisi TRAIN saja;
val/test (split resmi ISIC2017) tidak pernah diubah. Hasil dari bagian ini
menentukan keanggotaan train/val/test set SECARA FINAL -- seed pada
eksperimen multi-seed di bawah TIDAK PERNAH memengaruhi hasil bagian ini.
"""

all_df = pd.concat([
    train_df.assign(split='train'),
    val_df.assign(split='val'),
    test_df.assign(split='test')
], ignore_index=True)

t_dataprep_start = time.time()
print(f'Total gambar yang dicek leakage: {len(all_df):,}')
print(all_df.groupby(['split', 'source']).size())

# ============================================================
# SKENARIO 1: SPARK DATA MANAGEMENT
# Menggunakan PySpark untuk Distributed Feature Extraction (MD5 & pHash)
# ============================================================
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, udf
from pyspark.sql.types import StringType

print('\nInisialisasi Spark Session untuk Data Management...')
spark = SparkSession.builder \
    .appName("SkinLesionDataManagement") \
    .config("spark.master", "local[*]") \
    .config("spark.driver.memory", "8g") \
    .getOrCreate()

def compute_file_hash(path, algo='md5', block_size=65536):
    h = hashlib.new(algo)
    try:
        with open(path, 'rb') as f:
            for chunk in iter(lambda: f.read(block_size), b''):
                h.update(chunk)
        return h.hexdigest()
    except Exception as e:
        return None

def compute_phash(path, hash_size=8):
    try:
        img = Image.open(path).convert('RGB')
        return str(imagehash.phash(img, hash_size=hash_size))
    except Exception as e:
        return None

compute_file_hash_udf = udf(compute_file_hash, StringType())
compute_phash_udf = udf(compute_phash, StringType())

print('Mengkonversi data ke Spark DataFrame...')
spark_all_df = spark.createDataFrame(all_df)

print('Menghitung file hash (MD5) dan perceptual hash (pHash) terdistribusi via PySpark...')
spark_all_df = spark_all_df.withColumn('file_hash', compute_file_hash_udf(col('image_path')))
spark_all_df = spark_all_df.withColumn('phash', compute_phash_udf(col('image_path')))

print('Ekstraksi fitur selesai. Mengonversi kembali ke Pandas...')
all_df = spark_all_df.toPandas()

print('Spark Data Management selesai. Menghentikan SparkSession...')
spark.stop()
# ============================================================

n_failed = all_df['file_hash'].isna().sum()
if n_failed > 0:
    print(f'{n_failed} file gagal dibaca/di-hash (akan diabaikan dari pengecekan).')

hash_valid = all_df.dropna(subset=['file_hash'])
dup_mask   = hash_valid.duplicated(subset='file_hash', keep=False)
dup_exact  = hash_valid[dup_mask].sort_values('file_hash')

print(f'\nRingkasan Exact Duplicate:')
print(f'   Grup hash duplikat : {dup_exact["file_hash"].nunique()}')
print(f'   Total file terlibat: {len(dup_exact)}')

leak_split_groups = dup_exact.groupby('file_hash')['split'].nunique()
leak_split_ids     = leak_split_groups[leak_split_groups > 1].index
exact_leak_split   = dup_exact[dup_exact['file_hash'].isin(leak_split_ids)]

print(f'\nExact duplicate LINTAS SPLIT (train/val/test): {len(leak_split_ids)} grup, '
      f'{len(exact_leak_split)} file')

leak_source_groups = dup_exact.groupby('file_hash')['source'].nunique()
leak_source_ids     = leak_source_groups[leak_source_groups > 1].index
exact_leak_source   = dup_exact[dup_exact['file_hash'].isin(leak_source_ids)]

print(f'\nExact duplicate LINTAS DATASET (ISIC2017 / HAM10000 / ISIC MSK): '
      f'{len(leak_source_ids)} grup, {len(exact_leak_source)} file')

phash_valid = all_df.dropna(subset=['phash']).reset_index(drop=True)
print(f'Total gambar dengan pHash valid: {len(phash_valid):,}')

hash_ints = np.array([int(str(h), 16) for h in phash_valid['phash']], dtype=np.uint64)

POPCOUNT_TABLE_8 = np.array([bin(i).count('1') for i in range(256)], dtype=np.uint8)

def hamming_distance_matrix(chunk_ints, all_ints):
    """Hitung Hamming distance antara satu chunk hash terhadap seluruh hash lainnya."""
    xor = np.bitwise_xor(chunk_ints[:, None], all_ints[None, :])
    xor_bytes = xor.view(np.uint8).reshape(xor.shape[0], xor.shape[1], 8)
    return POPCOUNT_TABLE_8[xor_bytes].sum(axis=-1)

HAMMING_THRESHOLD = 5
CHUNK_SIZE        = 500

near_dup_pairs = []
n = len(hash_ints)
for start in range(0, n, CHUNK_SIZE):
    end   = min(start + CHUNK_SIZE, n)
    dist  = hamming_distance_matrix(hash_ints[start:end], hash_ints)
    for local_i in range(end - start):
        global_i = start + local_i
        close_js = np.where(dist[local_i] <= HAMMING_THRESHOLD)[0]
        for j in close_js:
            if j > global_i:
                near_dup_pairs.append((global_i, j, int(dist[local_i, j])))

print(f'\nTotal pasangan near-duplicate (Hamming distance <= {HAMMING_THRESHOLD}): '
      f'{len(near_dup_pairs):,}')

near_dup_df = pd.DataFrame(near_dup_pairs, columns=['idx_a', 'idx_b', 'hamming_distance'])
for suf in ['a', 'b']:
    near_dup_df[f'path_{suf}']   = near_dup_df[f'idx_{suf}'].map(phash_valid['image_path'])
    near_dup_df[f'source_{suf}'] = near_dup_df[f'idx_{suf}'].map(phash_valid['source'])
    near_dup_df[f'split_{suf}']  = near_dup_df[f'idx_{suf}'].map(phash_valid['split'])

near_dup_df['cross_split']  = near_dup_df['split_a']  != near_dup_df['split_b']
near_dup_df['cross_source'] = near_dup_df['source_a'] != near_dup_df['source_b']

print(f'\nNear-duplicate LINTAS SPLIT (train/val/test)                : '
      f'{near_dup_df["cross_split"].sum()} pasangan')
print(f'Near-duplicate LINTAS DATASET (ISIC2017/HAM10000/ISIC MSK) : '
      f'{near_dup_df["cross_source"].sum()} pasangan')

# --- Remediasi: buang leakage lintas split dari sisi TRAIN saja ---
val_test_paths = set(all_df.loc[all_df['split'].isin(['val', 'test']), 'image_path'])

leak_exact_train_idx = set(
    exact_leak_split.loc[exact_leak_split['split'] == 'train'].index
) if len(exact_leak_split) > 0 else set()

leak_near_train_paths = set()
if near_dup_df['cross_split'].any():
    cross = near_dup_df[near_dup_df['cross_split']]
    leak_near_train_paths |= set(cross.loc[cross['split_a'] == 'train', 'path_a'])
    leak_near_train_paths |= set(cross.loc[cross['split_b'] == 'train', 'path_b'])

leak_exact_train_paths = set(all_df.loc[list(leak_exact_train_idx), 'image_path']) \
    if leak_exact_train_idx else set()
train_paths_to_drop = leak_exact_train_paths | leak_near_train_paths

print(f'Baris train yang akan dibuang karena leakage ke val/test: {len(train_paths_to_drop)}')

train_df = train_df[~train_df['image_path'].isin(train_paths_to_drop)].reset_index(drop=True)

print('\n--- RINGKASAN DATASET SETELAH LEAKAGE CLEANUP ---')
print(f'Train : {len(train_df):5,} samples')
print(f'Val   : {len(val_df):5,} samples')
print(f'Test  : {len(test_df):5,} samples')

# --- Deduplikasi dalam-train (connected components / union-find) ---
class UnionFind:
    def __init__(self, items):
        self.parent = {x: x for x in items}
    def find(self, x):
        while self.parent[x] != x:
            self.parent[x] = self.parent[self.parent[x]]
            x = self.parent[x]
        return x
    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra != rb:
            self.parent[ra] = rb

train_paths_all = train_df['image_path'].tolist()
train_paths_set = set(train_paths_all)
uf = UnionFind(train_paths_all)

train_hash = hash_valid[(hash_valid['split'] == 'train') &
                         (hash_valid['image_path'].isin(train_paths_set))]
for _, grp in train_hash.groupby('file_hash'):
    paths = grp['image_path'].tolist()
    for p in paths[1:]:
        uf.union(paths[0], p)

train_near = near_dup_df[(near_dup_df['split_a'] == 'train') & (near_dup_df['split_b'] == 'train') &
                          (near_dup_df['path_a'].isin(train_paths_set)) &
                          (near_dup_df['path_b'].isin(train_paths_set))]
for _, row in train_near.iterrows():
    uf.union(row['path_a'], row['path_b'])

clusters = {}
for p in train_paths_all:
    root = uf.find(p)
    clusters.setdefault(root, []).append(p)

paths_to_keep = set()
paths_to_drop_intraset = set()
for root, members in clusters.items():
    members_sorted = sorted(members)
    paths_to_keep.add(members_sorted[0])
    paths_to_drop_intraset.update(members_sorted[1:])

n_clusters_with_dup = sum(1 for m in clusters.values() if len(m) > 1)
print(f'Grup duplikat/near-duplicate sesama train: {n_clusters_with_dup}')
print(f'Baris train yang dibuang (redundan, sisakan 1/grup): {len(paths_to_drop_intraset)}')

train_df = train_df[~train_df['image_path'].isin(paths_to_drop_intraset)].reset_index(drop=True)

# ------------------------------------------------------------------
# Freeze row order of every split NOW, before the seed loop begins. From
# this point on, train_df/val_df/test_df membership AND row order are
# FINAL and must never be touched again by any seed-dependent operation.
# ------------------------------------------------------------------
train_df = train_df.reset_index(drop=True)
val_df   = val_df.reset_index(drop=True)
test_df  = test_df.reset_index(drop=True)

print('\n--- RINGKASAN DATASET SETELAH DEDUP DALAM-TRAIN (FINAL, seed-independent) ---')
print(f'Train : {len(train_df):5,} samples')
print(f'Val   : {len(val_df):5,} samples')
print(f'Test  : {len(test_df):5,} samples')
print('\nDistribusi kelas train setelah dedup:')
print(train_df['dx'].value_counts())

"""## 4c. Pre-Experiment Sanity Checks

# ============================================================
# MULTI-SEED STABILITY IMPROVEMENT
# ============================================================
Hard assertions on the FROZEN train/val/test split, run once before any
seed-dependent code executes. The experiment stops immediately with an
informative error if any of these do not hold -- it never continues
silently on a violated assumption.
"""

assert train_df['image_path'].is_unique, 'train_df contains duplicate image_path after dedup.'
assert val_df['image_path'].is_unique, 'val_df contains duplicate image_path.'
assert test_df['image_path'].is_unique, 'test_df contains duplicate image_path.'

_train_paths = set(train_df['image_path'])
_val_paths   = set(val_df['image_path'])
_test_paths  = set(test_df['image_path'])
assert _train_paths.isdisjoint(_val_paths),  'train/val overlap detected by image_path!'
assert _train_paths.isdisjoint(_test_paths), 'train/test overlap detected by image_path!'
assert _val_paths.isdisjoint(_test_paths),   'val/test overlap detected by image_path!'

assert train_df['image_path'].notna().all(), 'train_df has null image_path.'
assert val_df['image_path'].notna().all(),   'val_df has null image_path.'
assert test_df['image_path'].notna().all(),  'test_df has null image_path.'

assert NUM_CLASSES == 3, f'Expected NUM_CLASSES == 3, got {NUM_CLASSES}.'
assert set(train_df['label'].unique()) <= {0, 1, 2}, 'train_df has label outside {0,1,2}.'
assert set(val_df['label'].unique())   <= {0, 1, 2}, 'val_df has label outside {0,1,2}.'
assert set(test_df['label'].unique())  <= {0, 1, 2}, 'test_df has label outside {0,1,2}.'

_missing_files = [
    p for p in pd.concat([train_df['image_path'], val_df['image_path'], test_df['image_path']])
    if not os.path.exists(p)
]
assert len(_missing_files) == 0, (
    f'{len(_missing_files)} image_path entries do not exist on disk, e.g. {_missing_files[:5]}'
)

print('Pre-experiment sanity checks passed: no train/val/test overlap, '
      'no null/missing image_path, labels in {0,1,2}.')
t_dataprep_end = time.time()
dataprep_duration_sec = t_dataprep_end - t_dataprep_start
print(f'\n>>> [{SCENARIO_LABEL}] WAKTU DATA MANAGEMENT (PySpark UDF MD5 + pHash + Leakage Cleanup + Dedup): '
      f'{dataprep_duration_sec:.2f} detik ({len(all_df)/dataprep_duration_sec:.1f} gambar/detik) <<<\n')

"""## 5. Class Distribution per Source (Diagnostic Chart)"""

DX_LABEL_NAME = {'nv': 'nevus', 'mel': 'melanoma', 'bkl': 'seborrheic_keratosis'}

pivot_source = (
    train_df
    .groupby(['source', 'dx'])
    .size()
    .unstack(fill_value=0)
)
pivot_source = pivot_source[['nv', 'mel', 'bkl']]
pivot_source.columns = [DX_LABEL_NAME[c] for c in pivot_source.columns]

fig, ax = plt.subplots(figsize=(9, 5.5))
pivot_source.plot(kind='bar', ax=ax, color=['#4CAF50', '#F44336', '#2196F3'], width=0.7)
ax.set_xlabel('Data Source')
ax.set_ylabel('Sample Count')
ax.legend(title='Class')
plt.xticks(rotation=15, ha='right')
for container in ax.containers:
    ax.bar_label(container, fontsize=9)
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/class_distribution_by_source.png', dpi=150, bbox_inches='tight')
plt.show()

"""## 6. Preprocessing & Dataset Definition

Transform dan augmentasi TIDAK DIUBAH dari pipeline asli. `HAMDataset`
mendapat SATU parameter opsional baru (`return_path`) -- lihat komentar
MULTI-SEED STABILITY IMPROVEMENT di bawah -- tanpa mengubah perilaku
default-nya (training loop tetap kompatibel, tetap menerima (image, label)).
"""

IMG_SIZE      = 224
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomRotation(30),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

val_test_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])


class HAMDataset(Dataset):
    def __init__(self, dataframe, transform=None, return_path=False):
        self.df        = dataframe
        self.transform = transform
        # ============================================================
        # MULTI-SEED STABILITY IMPROVEMENT
        # return_path=True lets evaluation carry an explicit image_path
        # identifier out of the DataLoader, so predictions can be aligned
        # across seeds by identifier (merge) instead of assumed row
        # position. Default stays False, so training/validation loaders
        # (train_one_epoch/evaluate_epoch, which unpack (image, label))
        # are completely unaffected.
        # ============================================================
        self.return_path = return_path

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row     = self.df.iloc[idx]
        label   = int(row['label'])
        img_bgr = cv2.imread(row['image_path'])
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
        img_pil = Image.fromarray(img_rgb)
        if self.transform:
            img_pil = self.transform(img_pil)
        if self.return_path:
            return img_pil, label, row['image_path']
        return img_pil, label


BATCH_SIZE  = 32
NUM_WORKERS = 2

print(f'Batch size : {BATCH_SIZE}')
print(f'Train rows : {len(train_df):,} | Val rows: {len(val_df):,} | Test rows: {len(test_df):,}')

"""## 7. CNN Model (EfficientNet-B0)

Arsitektur TIDAK DIUBAH dari pipeline asli. Instansiasi model dipindah ke
fungsi `create_model(seed)` di bagian eksperimen multi-seed, karena model
harus dibuat ulang dari awal (fresh classifier-head init) pada setiap seed.
"""

class EfficientNetB0Classifier(nn.Module):
    def __init__(self, num_classes, pretrained=True):
        super().__init__()
        self.backbone = timm.create_model('efficientnet_b0', pretrained=pretrained, num_classes=0)
        feature_dim = self.backbone.num_features  # 1280
        self.head = nn.Sequential(
            nn.Dropout(p=0.4),
            nn.Linear(feature_dim, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(p=0.3),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        features = self.backbone(x)
        return self.head(features)


# Sanity check of output shape (does not consume/affect any seed's RNG
# state because it runs once, before the seed loop, using a fresh
# throwaway model that is discarded immediately after).
_sanity_model = EfficientNetB0Classifier(NUM_CLASSES, pretrained=True).to(DEVICE)
_dummy = torch.randn(2, 3, IMG_SIZE, IMG_SIZE).to(DEVICE)
with torch.no_grad():
    _out = _sanity_model(_dummy)
print(f'EfficientNet-B0 output shape     : {_out.shape}')
print(f'EfficientNet-B0 trainable params : {sum(p.numel() for p in _sanity_model.parameters() if p.requires_grad):,}')
del _sanity_model, _dummy, _out
torch.cuda.empty_cache()

"""## 8. Loss Function (Focal Loss)"""

class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0, reduction='mean'):
        super().__init__()
        self.gamma     = gamma
        self.reduction = reduction
        self.ce        = nn.CrossEntropyLoss(reduction='none')

    def forward(self, logits, targets):
        ce_loss    = self.ce(logits, targets)
        p_t        = torch.exp(-ce_loss)
        focal_loss = (1 - p_t) ** self.gamma * ce_loss
        return focal_loss.mean() if self.reduction == 'mean' else focal_loss.sum()


FOCAL_GAMMA          = 2.0
LEARNING_RATE        = 1e-4
MAX_EPOCHS           = 10
EARLY_STOP_PATIENCE  = 5
EARLY_STOP_MIN_DELTA = 1e-4

print(f'Loss = FocalLoss(gamma={FOCAL_GAMMA}), Optimizer = AdamW(lr={LEARNING_RATE})')

"""## 9. Multi-Seed Reproducibility Infrastructure

`SEEDS` adalah lima controlled random seed yang telah ditentukan. Setiap
seed adalah satu independent training run: model & optimizer dibuat ulang
dari awal, dataset/split TIDAK berubah, dan seed hanya memengaruhi proses
stokastik (init classifier head, augmentasi, weighted sampling, urutan
mini-batch, dropout).
"""

SEEDS = [42, 123, 2026, 3407, 8901]


def set_seed(seed):
    """Set semua sumber randomness Python/NumPy/PyTorch(CPU+CUDA) dan
    aktifkan mode deterministik cuDNN/PyTorch untuk seed yang diberikan.

    Catatan: pengaturan ini MEMINIMALKAN uncontrolled nondeterminism di
    dalam satu environment eksperimen yang tetap -- ini BUKAN jaminan
    reproducibility absolut lintas hardware, versi CUDA, versi PyTorch,
    atau sistem operasi yang berbeda."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    # ============================================================
    # MULTI-SEED STABILITY IMPROVEMENT: eksplisit manual_seed (device
    # tunggal) + manual_seed_all (semua device CUDA) -- keduanya diset
    # untuk redundansi seperti diminta pada review spec.
    # ============================================================
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
    try:
        torch.use_deterministic_algorithms(True, warn_only=True)
    except TypeError:
        # PyTorch lama tidak punya argumen warn_only
        torch.use_deterministic_algorithms(True)


def seed_worker(worker_id):
    """worker_init_fn untuk DataLoader: derive seed tiap worker dari
    torch.initial_seed() (yang sudah di-set lewat generator DataLoader),
    supaya augmentasi tiap worker reproducible per seed."""
    worker_seed = torch.initial_seed() % (2 ** 32)
    np.random.seed(worker_seed)
    random.seed(worker_seed)


def make_weighted_sampler(labels, generator=None):
    """WeightedRandomSampler dengan class weight = 1/frekuensi kelas, dan
    generator eksplisit supaya urutan sampling reproducible per seed."""
    class_counts   = np.bincount(labels)
    class_weights  = 1.0 / class_counts
    sample_weights = class_weights[labels]
    return WeightedRandomSampler(
        weights=torch.DoubleTensor(sample_weights),
        num_samples=len(sample_weights),
        replacement=True,
        generator=generator,
    )


def create_dataloaders(seed):
    """Buat ulang train/val/test DataLoader untuk satu seed. Dataset
    (train_df/val_df/test_df) SELALU identik -- hanya randomness sampling,
    urutan mini-batch, dan augmentasi worker yang di-seed berbeda per seed.

    # ============================================================
    # MULTI-SEED STABILITY IMPROVEMENT
    # ============================================================
    WeightedRandomSampler dan setiap DataLoader sekarang memakai
    torch.Generator TERPISAH (bukan satu generator yang sama dipakai
    ulang), dengan offset besar (+10000/+20000/+30000) supaya stream
    RNG satu seed tidak pernah bertumpang tindih dengan stream RNG seed
    lain. test_dataset memakai return_path=True supaya evaluate_model bisa
    membawa image_path eksplisit per sampel (lihat Section 3 review spec).
    """
    sampler_generator = torch.Generator()
    sampler_generator.manual_seed(seed)

    train_loader_generator = torch.Generator()
    train_loader_generator.manual_seed(seed + 10000)

    val_loader_generator = torch.Generator()
    val_loader_generator.manual_seed(seed + 20000)

    test_loader_generator = torch.Generator()
    test_loader_generator.manual_seed(seed + 30000)

    train_dataset = HAMDataset(train_df, transform=train_transform)
    val_dataset   = HAMDataset(val_df,   transform=val_test_transform)
    test_dataset  = HAMDataset(test_df,  transform=val_test_transform, return_path=True)

    train_sampler = make_weighted_sampler(train_df['label'].values, generator=sampler_generator)

    train_loader = DataLoader(
        train_dataset, batch_size=BATCH_SIZE, sampler=train_sampler,
        num_workers=NUM_WORKERS, pin_memory=True,
        worker_init_fn=seed_worker, generator=train_loader_generator,
    )
    val_loader = DataLoader(
        val_dataset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, pin_memory=True,
        worker_init_fn=seed_worker, generator=val_loader_generator,
    )
    test_loader = DataLoader(
        test_dataset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, pin_memory=True,
        worker_init_fn=seed_worker, generator=test_loader_generator,
    )
    return train_loader, val_loader, test_loader


def create_model(seed):
    """Buat model EfficientNet-B0 dari awal. set_seed(seed) dipanggil tepat
    sebelum instansiasi sehingga init classifier-head (Linear/BatchNorm)
    ter-seed secara deterministik per seed; backbone pretrained ImageNet
    tetap identik (tidak dipengaruhi seed)."""
    set_seed(seed)
    model = EfficientNetB0Classifier(NUM_CLASSES, pretrained=True).to(DEVICE)
    return model


class EarlyStopping:
    def __init__(self, patience=5, min_delta=1e-4, mode='max'):
        self.patience, self.min_delta, self.mode = patience, min_delta, mode
        self.counter, self.best, self.stop = 0, (-np.inf if mode == 'max' else np.inf), False

    def __call__(self, metric):
        improved = (metric > self.best + self.min_delta) if self.mode == 'max' else (metric < self.best - self.min_delta)
        if improved:
            self.best, self.counter = metric, 0
        else:
            self.counter += 1
            self.stop = self.counter >= self.patience
        return self.stop


def train_one_epoch(model, loader, optimiser, criterion, device):
    model.train()
    total_loss, correct, total = 0.0, 0, 0
    for images, labels in loader:
        images, labels = images.to(device), labels.to(device)
        optimiser.zero_grad()
        logits = model(images)
        loss   = criterion(logits, labels)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimiser.step()
        total_loss += loss.item() * images.size(0)
        correct    += (logits.argmax(1) == labels).sum().item()
        total      += images.size(0)
    return total_loss / total, correct / total


@torch.no_grad()
def evaluate_epoch(model, loader, criterion, device, num_classes):
    """Evaluasi cepat (loss/acc/macro-AUC) dipakai di dalam training loop
    untuk validation-based model selection & early stopping."""
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    all_probs, all_labels = [], []
    for images, labels in loader:
        images, labels = images.to(device), labels.to(device)
        logits = model(images)
        loss   = criterion(logits, labels)
        total_loss += loss.item() * images.size(0)
        probs = torch.softmax(logits, dim=1)
        correct += (probs.argmax(1) == labels).sum().item()
        total   += images.size(0)
        all_probs.append(probs.cpu().numpy())
        all_labels.append(labels.cpu().numpy())
    all_probs  = np.concatenate(all_probs)
    all_labels = np.concatenate(all_labels)
    labels_bin = label_binarize(all_labels, classes=list(range(num_classes)))
    macro_auc  = roc_auc_score(labels_bin, all_probs, multi_class='ovr', average='macro')
    return total_loss / total, correct / total, macro_auc


def train_model(model, train_loader, val_loader, criterion, optimiser, device,
                 num_classes, seed, num_epochs=MAX_EPOCHS, patience=EARLY_STOP_PATIENCE,
                 model_name='model', history_tag=None):
    """Training loop untuk satu run. Model selection & early stopping
    berbasis validation Macro-AUC (identik konfigurasinya untuk semua
    seed). Test set TIDAK PERNAH dipakai di sini.

    # ============================================================
    # MULTI-SEED STABILITY IMPROVEMENT
    # ============================================================
    `history_tag` memisahkan nama file training_history_*.csv dari
    `model_name`/`seed`, supaya repeatability check opsional (seed 42
    dijalankan dua kali, Section 9b) bisa memakai fungsi ini tanpa menimpa
    training_history_seed42.csv resmi dari eksperimen utama 5-seed."""
    if history_tag is None:
        history_tag = f'seed{seed}'

    early_stopper = EarlyStopping(patience=patience, min_delta=EARLY_STOP_MIN_DELTA, mode='max')
    history = {'epoch': [], 'train_loss': [], 'train_acc': [],
               'val_loss': [], 'val_acc': [], 'val_auc': []}
    best_auc, best_state, best_epoch = -np.inf, None, None
    best_loss, best_acc = None, None

    t_start = time.time()
    for epoch in range(1, num_epochs + 1):
        t0 = time.time()
        tr_loss, tr_acc = train_one_epoch(model, train_loader, optimiser, criterion, device)
        val_loss, val_acc, val_auc = evaluate_epoch(model, val_loader, criterion, device, num_classes)

        history['epoch'].append(epoch)
        history['train_loss'].append(tr_loss); history['train_acc'].append(tr_acc)
        history['val_loss'].append(val_loss);   history['val_acc'].append(val_acc)
        history['val_auc'].append(val_auc)

        if val_auc > best_auc:
            best_auc, best_loss, best_acc, best_epoch = val_auc, val_loss, val_acc, epoch
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}

        print(f'[{history_tag}] Epoch [{epoch:02d}/{num_epochs}] '
              f'Train Loss: {tr_loss:.4f} Acc: {tr_acc:.4f} | '
              f'Val Loss: {val_loss:.4f} Acc: {val_acc:.4f} AUC: {val_auc:.4f} '
              f'[{time.time()-t0:.0f}s]')

        if early_stopper(val_auc):
            print(f'[{history_tag}][Early Stopping] tidak ada peningkatan selama {patience} epoch.')
            break

    training_time_sec = time.time() - t_start
    model.load_state_dict(best_state)

    ckpt_path = os.path.join(OUTPUT_DIR, f'{model_name}_best.pth')
    torch.save(best_state, ckpt_path)

    hist_df = pd.DataFrame(history)
    hist_path = os.path.join(OUTPUT_DIR, f'training_history_{history_tag}.csv')
    hist_df.to_csv(hist_path, index=False)

    return {
        'history': history, 'best_loss': best_loss, 'best_acc': best_acc,
        'best_auc': best_auc, 'best_epoch': best_epoch,
        'training_time_sec': training_time_sec,
        'checkpoint_path': ckpt_path, 'history_path': hist_path,
    }


def evaluate_model(model, loader, df_reference, criterion, device, num_classes, class_names):
    """Evaluasi LENGKAP satu kali di test set (dipanggil sekali per run,
    setelah training selesai, memakai best checkpoint run tsb).

    # ============================================================
    # MULTI-SEED STABILITY IMPROVEMENT
    # ============================================================
    Setiap prediksi sekarang membawa image_path-nya sendiri langsung dari
    DataLoader (HAMDataset(..., return_path=True)) -- BUKAN diasumsikan
    dari urutan baris df_reference. df_reference tetap dipakai untuk
    memvalidasi kelengkapan (jumlah sama, set identifier sama, true_label
    konsisten) -- bukan untuk MENYEDIAKAN identifier. Juga menambahkan
    sanity checks: jumlah prediksi == jumlah test sample, image_path unik,
    shape probability == (N,3), semua probability di [0,1], dan jumlah
    probability tiap gambar mendekati 1 -- gagal dengan error informatif
    jika salah satu dilanggar."""
    model.eval()
    total_loss, total = 0.0, 0
    all_probs, all_preds, all_labels, all_paths = [], [], [], []
    with torch.no_grad():
        for images, labels, paths in loader:
            images_dev, labels_dev = images.to(device), labels.to(device)
            logits = model(images_dev)
            loss   = criterion(logits, labels_dev)
            total_loss += loss.item() * images.size(0)
            total      += images.size(0)
            probs = torch.softmax(logits, dim=1)
            preds = probs.argmax(dim=1)
            all_probs.append(probs.cpu().numpy())
            all_preds.append(preds.cpu().numpy())
            all_labels.append(labels.numpy())
            all_paths.extend(list(paths))

    all_probs  = np.concatenate(all_probs)
    all_preds  = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)

    per_image_df = pd.DataFrame({
        'image_path': all_paths,
        'true_label': all_labels,
        'pred_label': all_preds,
    })
    for i, cname in enumerate(class_names):
        per_image_df[f'prob_{cname}'] = all_probs[:, i]

    # ---- sanity checks (fail loudly) ----
    assert len(per_image_df) == len(df_reference), (
        f'evaluate_model: got {len(per_image_df)} predictions, expected '
        f'{len(df_reference)} test samples.'
    )
    assert per_image_df['image_path'].is_unique, (
        'evaluate_model: duplicate image_path among predictions -- each test image '
        'must appear exactly once.'
    )
    assert set(per_image_df['image_path']) == set(df_reference['image_path']), (
        'evaluate_model: set of predicted image_path does not match df_reference -- '
        'missing or unexpected test images.'
    )
    ref_label_by_path = df_reference.set_index('image_path')['label']
    mismatched = per_image_df[per_image_df['image_path'].map(ref_label_by_path) != per_image_df['true_label']]
    assert len(mismatched) == 0, (
        f'evaluate_model: true_label mismatch vs df_reference for {len(mismatched)} image(s).'
    )
    assert all_probs.shape == (len(df_reference), num_classes), (
        f'evaluate_model: probability array shape {all_probs.shape} != '
        f'expected {(len(df_reference), num_classes)}.'
    )
    assert np.all((all_probs >= 0.0) & (all_probs <= 1.0)), (
        'evaluate_model: some predicted probabilities fall outside [0, 1].'
    )
    assert np.allclose(all_probs.sum(axis=1), 1.0, atol=1e-3), (
        'evaluate_model: predicted class probabilities do not sum to ~1 for some images.'
    )

    test_loss = total_loss / total
    accuracy  = accuracy_score(all_labels, all_preds)

    precision, recall, f1, _ = precision_recall_fscore_support(
        all_labels, all_preds, labels=list(range(num_classes)), zero_division=0
    )
    macro_precision, macro_recall, macro_f1 = precision.mean(), recall.mean(), f1.mean()

    labels_bin = label_binarize(all_labels, classes=list(range(num_classes)))
    macro_auc = roc_auc_score(labels_bin, all_probs, multi_class='ovr', average='macro')

    classwise = []
    for i, cname in enumerate(class_names):
        try:
            auc_i = roc_auc_score(labels_bin[:, i], all_probs[:, i])
        except ValueError:
            auc_i = np.nan
        classwise.append({
            'class_name': cname, 'precision': precision[i], 'recall': recall[i],
            'f1': f1[i], 'auc': auc_i,
        })

    return {
        'test_loss': test_loss, 'accuracy': accuracy,
        'macro_precision': macro_precision, 'macro_recall': macro_recall,
        'macro_f1': macro_f1, 'macro_auc': macro_auc,
        'classwise': classwise, 'per_image': per_image_df,
        'confusion_matrix': confusion_matrix(all_labels, all_preds),
    }


def run_single_seed_training(seed, model_name, history_tag):
    """
    # ============================================================
    # MULTI-SEED STABILITY IMPROVEMENT
    # ============================================================
    Shared execution path for ONE independent run: reseed everything,
    build fresh DataLoaders/model/optimizer, train with early stopping on
    validation Macro-AUC, then evaluate the best checkpoint once on the
    fixed test set. Used identically by the main 5-seed experiment
    (Section 10) and by the optional seed-42 repeatability check (Section
    9b), so both paths cannot silently diverge in behaviour."""
    set_seed(seed)
    train_loader, val_loader, test_loader = create_dataloaders(seed)
    model     = create_model(seed)
    optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE)
    criterion = FocalLoss(gamma=FOCAL_GAMMA)

    train_result = train_model(
        model, train_loader, val_loader, criterion, optimizer, DEVICE, NUM_CLASSES, seed,
        num_epochs=MAX_EPOCHS, patience=EARLY_STOP_PATIENCE,
        model_name=model_name, history_tag=history_tag,
    )
    t_inf_start = time.time()
    test_result = evaluate_model(
        model, test_loader, test_df, criterion, DEVICE, NUM_CLASSES, CLASS_NAMES
    )
    t_inf_end = time.time()
    test_result['inference_time_sec'] = t_inf_end - t_inf_start
    test_result['peak_gpu_mem_mb'] = (
        torch.cuda.max_memory_allocated() / (1024**2) if torch.cuda.is_available() else 0.0
    )

    del train_loader, val_loader, test_loader, optimizer
    torch.cuda.empty_cache()

    return model, train_result, test_result


def save_environment_info(stage='START'):
    """
    # ============================================================
    # MULTI-SEED STABILITY IMPROVEMENT
    # ============================================================
    Print AND append an environment/configuration snapshot to
    experiment_environment.txt. Does not claim exact bitwise
    reproducibility across different hardware/CUDA/PyTorch/OS -- only
    that uncontrolled nondeterminism is minimized within one fixed
    experimental environment."""
    lines = []
    lines.append('=' * 70)
    lines.append(f' ENVIRONMENT / REPRODUCIBILITY INFO ({stage}) '.center(70, '='))
    lines.append('=' * 70)
    lines.append(f'Experiment timestamp : {datetime.now().isoformat(timespec="seconds")}')
    lines.append(f'Python version       : {sys.version.split()[0]}')
    lines.append(f'Platform             : {platform.platform()}')
    lines.append(f'PyTorch version      : {torch.__version__}')
    lines.append(f'torchvision version  : {torchvision.__version__}')
    lines.append(f'timm version         : {timm.__version__}')
    lines.append(f'CUDA available       : {torch.cuda.is_available()}')
    if torch.cuda.is_available():
        lines.append(f'CUDA version         : {torch.version.cuda}')
        lines.append(f'cuDNN version        : {torch.backends.cudnn.version()}')
        lines.append(f'GPU name             : {torch.cuda.get_device_name(0)}')
    lines.append(f'NUM_WORKERS          : {NUM_WORKERS}')
    lines.append(f'Batch size           : {BATCH_SIZE}')
    lines.append(f'Learning rate        : {LEARNING_RATE}')
    lines.append(f'Optimizer            : AdamW')
    lines.append(f'Loss function        : FocalLoss (gamma={FOCAL_GAMMA})')
    lines.append(f'Max epochs           : {MAX_EPOCHS}')
    lines.append(f'Early stop patience  : {EARLY_STOP_PATIENCE}')
    lines.append(f'Image size           : {IMG_SIZE}x{IMG_SIZE}')
    lines.append(f'Controlled seeds     : {SEEDS}')
    lines.append(f'DATA_PREP_SEED       : {DATA_PREP_SEED}')
    lines.append(f'Train samples        : {len(train_df):,}')
    lines.append(f'Val samples          : {len(val_df):,}')
    lines.append(f'Test samples         : {len(test_df):,}')
    lines.append('Train class distribution (0=nevus,1=melanoma,2=seb.keratosis):')
    lines.append(train_df['label'].value_counts().sort_index().to_string())
    lines.append('Val class distribution:')
    lines.append(val_df['label'].value_counts().sort_index().to_string())
    lines.append('Test class distribution:')
    lines.append(test_df['label'].value_counts().sort_index().to_string())
    lines.append('=' * 70)
    lines.append('')

    text = '\n'.join(lines)
    print(text)
    with open(os.path.join(OUTPUT_DIR, 'experiment_environment.txt'), 'a', encoding='utf-8') as f:
        f.write(text + '\n')


save_environment_info('START')

"""## 9b. Optional Repeatability Check (Seed 42, Diagnostic Only)

# ============================================================
# MULTI-SEED STABILITY IMPROVEMENT
# ============================================================
Runs seed 42 TWICE, independently, with identical configuration, purely as
a diagnostic check of within-environment repeatability. This is NOT one of
the five main observations -- Run A and Run B are NEVER mixed into the
mean +/- SD statistics computed from SEEDS = [42, 123, 2026, 3407, 8901].
Set RUN_REPEATABILITY_CHECK = False to skip (saves two extra full
training runs).
"""

RUN_REPEATABILITY_CHECK = True
REPEATABILITY_SEED = 42

if RUN_REPEATABILITY_CHECK:
    print('\n' + '=' * 70)
    print(f' REPEATABILITY CHECK: seed {REPEATABILITY_SEED}, Run A vs Run B (diagnostic only) '.center(70, '='))
    print('=' * 70)

    repeat_rows = []
    repeat_per_image = {}
    for run_label in ['A', 'B']:
        _repeat_model, train_result_r, test_result_r = run_single_seed_training(
            REPEATABILITY_SEED,
            model_name=f'efficientnet_b0_seed{REPEATABILITY_SEED}_repeat{run_label}',
            history_tag=f'seed{REPEATABILITY_SEED}_repeat{run_label}',
        )
        repeat_rows.append({
            'seed': REPEATABILITY_SEED,
            'repetition': run_label,
            'best_epoch': train_result_r['best_epoch'],
            'best_val_auc': train_result_r['best_auc'],
            'test_accuracy': test_result_r['accuracy'],
            'macro_precision': test_result_r['macro_precision'],
            'macro_recall': test_result_r['macro_recall'],
            'macro_f1': test_result_r['macro_f1'],
            'macro_auc': test_result_r['macro_auc'],
            'training_time': train_result_r['training_time_sec'],
        })
        repeat_per_image[run_label] = test_result_r['per_image']
        del _repeat_model
        torch.cuda.empty_cache()

    reproducibility_df = pd.DataFrame(repeat_rows)

    _prob_cols = [f'prob_{c}' for c in CLASS_NAMES]
    _run_a = repeat_per_image['A'][['image_path', 'true_label', 'pred_label'] + _prob_cols]
    _run_b = repeat_per_image['B'][['image_path', 'true_label', 'pred_label'] + _prob_cols]
    _merged_ab = _run_a.merge(_run_b, on='image_path', suffixes=('_A', '_B'))
    assert len(_merged_ab) == len(test_df), (
        'Repeatability check: Run A/Run B prediction merge row count mismatch.'
    )
    assert (_merged_ab['true_label_A'] == _merged_ab['true_label_B']).all(), (
        'Repeatability check: true_label differs between Run A and Run B for the same image_path.'
    )

    n_diff_preds = int((_merged_ab['pred_label_A'] != _merged_ab['pred_label_B']).sum())
    max_abs_prob_diff = float(max(
        (_merged_ab[f'prob_{c}_A'] - _merged_ab[f'prob_{c}_B']).abs().max() for c in CLASS_NAMES
    ))

    reproducibility_df['number_of_different_test_predictions'] = n_diff_preds
    reproducibility_df['maximum_absolute_probability_difference'] = max_abs_prob_diff
    reproducibility_df.to_csv(os.path.join(OUTPUT_DIR, 'reproducibility_check.csv'), index=False)

    print(f'Run A vs Run B: {n_diff_preds}/{len(_merged_ab)} test predictions differ; '
          f'max |probability difference| = {max_abs_prob_diff:.6f}')
    print('reproducibility_check.csv saved. (Diagnostic only -- NOT part of the 5 main seeds.)')
else:
    print('RUN_REPEATABILITY_CHECK = False -- skipping seed-42 repeatability diagnostic.')

"""## 10. Run Controlled Multi-Seed Experiment

Loop utama: untuk setiap seed di `SEEDS`, model & optimizer dibuat ulang
dari awal, DataLoader dibuat ulang dengan randomness ter-seed, training
dijalankan dengan early stopping berbasis validation Macro-AUC, lalu best
model dievaluasi SATU KALI di test set yang tetap. Test set TIDAK PERNAH
dipakai untuk early stopping/hyperparameter tuning/model selection/pemilihan
seed.
"""

overall_rows      = []
classwise_rows    = []
run_manifest_rows = []
run_artifacts      = {}  # seed -> {'checkpoint_path', 'per_image', 'confusion_matrix'}

for seed in SEEDS:
    print('\n' + '=' * 70)
    print(f' SEED = {seed} '.center(70, '='))
    print('=' * 70)

    model, train_result, test_result = run_single_seed_training(
        seed, model_name=f'efficientnet_b0_seed{seed}', history_tag=f'seed{seed}'
    )

    overall_rows.append({
        'scenario': SCENARIO_ID,
        'seed': seed,
        'test_loss': test_result['test_loss'],
        'accuracy': test_result['accuracy'],
        'macro_precision': test_result['macro_precision'],
        'macro_recall': test_result['macro_recall'],
        'macro_f1': test_result['macro_f1'],
        'macro_auc': test_result['macro_auc'],
        'best_epoch': train_result['best_epoch'],
        'best_val_auc': train_result['best_auc'],
        'training_time_sec': train_result['training_time_sec'],
        'inference_time_sec': test_result.get('inference_time_sec', 0.0),
        'peak_gpu_mem_mb': test_result.get('peak_gpu_mem_mb', 0.0),
    })

    for cw in test_result['classwise']:
        classwise_rows.append({'seed': seed, **cw})

    run_artifacts[seed] = {
        'checkpoint_path': train_result['checkpoint_path'],
        'per_image': test_result['per_image'],
        'confusion_matrix': test_result['confusion_matrix'],
    }

    run_manifest_rows.append({
        'seed': seed,
        'checkpoint_path': train_result['checkpoint_path'],
        'history_path': train_result['history_path'],
        'best_epoch': train_result['best_epoch'],
        'best_val_auc': train_result['best_auc'],
        'test_accuracy': test_result['accuracy'],
        'macro_f1': test_result['macro_f1'],
        'macro_auc': test_result['macro_auc'],
        'training_time': train_result['training_time_sec'],
        'status': 'success',
    })

    print(f'[Seed {seed}] TEST -> Loss: {test_result["test_loss"]:.4f} | '
          f'Acc: {test_result["accuracy"]:.4f} | Macro-F1: {test_result["macro_f1"]:.4f} | '
          f'Macro-AUC: {test_result["macro_auc"]:.4f}')

    del model
    torch.cuda.empty_cache()

overall_df   = pd.DataFrame(overall_rows)
classwise_df = pd.DataFrame(classwise_rows)

run_manifest_df = pd.DataFrame(run_manifest_rows)
run_manifest_df.to_csv(os.path.join(OUTPUT_DIR, 'run_manifest.csv'), index=False)
print('run_manifest.csv saved.')

print('\n' + '=' * 70)
print(' PER-SEED OVERALL TEST RESULTS (no seed is treated as "the" result) ')
print('=' * 70)
print(overall_df.to_string(index=False))

"""## 11. Aggregate Multi-Seed Statistics (Mean +/- SD, CV, 95% CI)

Statistik dihitung dari kelima seed SECARA AGREGAT. Tidak ada seed
terbaik yang dipilih sebagai hasil utama -- hasil utama adalah mean +/- SD
(dan CV, 95% CI) di seluruh overall metric dan class-wise metric.
"""

def summarize_metric(values):
    """Mean, sample SD (ddof=1), min, max, range, CV(%), dan 95% CI (t-dist,
    df=n-1) untuk sekumpulan nilai metrik dari n seed."""
    values = np.asarray(values, dtype=float)
    n = len(values)
    mean = values.mean()
    sd = values.std(ddof=1) if n > 1 else 0.0
    vmin, vmax = values.min(), values.max()
    vrange = vmax - vmin
    cv_pct = (sd / mean * 100) if mean != 0 else np.nan
    if n > 1 and sd > 0:
        t_crit = sp_stats.t.ppf(0.975, df=n - 1)
        ci_half = t_crit * sd / math.sqrt(n)
    else:
        ci_half = 0.0
    return {
        'mean': mean, 'sd': sd, 'min': vmin, 'max': vmax, 'range': vrange,
        'cv_pct': cv_pct, 'ci95_lower': mean - ci_half, 'ci95_upper': mean + ci_half,
        'n_seeds': n,
    }


def calculate_multiseed_statistics(overall_df, classwise_df, class_names):
    rows = []
    overall_metrics = ['test_loss', 'accuracy', 'macro_precision',
                        'macro_recall', 'macro_f1', 'macro_auc']
    for m in overall_metrics:
        stat = summarize_metric(overall_df[m].values)
        rows.append({'scope': 'overall', 'metric': m, 'class_name': '-', **stat})

    for cname in class_names:
        sub = classwise_df[classwise_df['class_name'] == cname]
        for m in ['precision', 'recall', 'f1', 'auc']:
            stat = summarize_metric(sub[m].values)
            rows.append({'scope': 'classwise', 'metric': m, 'class_name': cname, **stat})

    return pd.DataFrame(rows)


summary_df = calculate_multiseed_statistics(overall_df, classwise_df, CLASS_NAMES)

print('\n' + '=' * 70)
print(' AGGREGATE RESULTS ACROSS 5 CONTROLLED SEEDS (Mean +/- SD) ')
print('=' * 70)
for m in ['accuracy', 'macro_precision', 'macro_recall', 'macro_f1', 'macro_auc']:
    row = summary_df[(summary_df['scope'] == 'overall') & (summary_df['metric'] == m)].iloc[0]
    if m == 'accuracy':
        print(f'Accuracy   = {row["mean"]*100:.2f} +/- {row["sd"]*100:.2f}%  '
              f'(min {row["min"]*100:.2f}%, max {row["max"]*100:.2f}%, CV={row["cv_pct"]:.2f}%)')
    else:
        print(f'{m:<16} = {row["mean"]:.4f} +/- {row["sd"]:.4f}  '
              f'(min {row["min"]:.4f}, max {row["max"]:.4f}, CV={row["cv_pct"]:.2f}%)')

print('\n--- Class-wise F1-score stability (multiseed_summary.csv) ---')
print(f'{"Class":<25}{"Mean F1":>10}{"SD":>10}{"CV(%)":>10}{"Min":>10}{"Max":>10}')
for cname in CLASS_NAMES:
    r = summary_df[(summary_df['scope'] == 'classwise') &
                    (summary_df['metric'] == 'f1') & (summary_df['class_name'] == cname)].iloc[0]
    print(f'{cname:<25}{r["mean"]:>10.4f}{r["sd"]:>10.4f}{r["cv_pct"]:>10.2f}{r["min"]:>10.4f}{r["max"]:>10.4f}')

print('\nNote: which class is "least stable" should be judged from SD, range, and CV above')
print('TOGETHER WITH prediction agreement and prediction entropy (Section 12 below) -- never')
print('from CV in isolation.')

"""## 12. Prediction Stability Across Seeds (Alignment, SAR, Entropy,
Probability Stability, Correctness Stability)

Untuk setiap gambar test: gabungkan prediksi kelima seed lewat MERGE pada
`image_path` (bukan asumsi posisi baris), lalu hitung Seed Agreement Rate
(SAR), majority prediction (eksplisit "tie" untuk kasus mis. 2-2-1),
kategori stabilitas 4-level, prediction entropy, probability-based
stability, dan correctness stability.
"""

def compute_agreement_and_majority(row, pred_cols, class_names, K):
    """Section 6: SAR, majority prediction (eksplisit 'tie' bila tidak ada
    kelas dengan pluralitas tunggal -- mis. split 2-2-1 di antara 3 kelas),
    dan kategori stabilitas 4-level."""
    preds = row[pred_cols].values.astype(int)
    counts = np.bincount(preds, minlength=len(class_names))
    max_count = int(counts.max())
    n_at_max = int((counts == max_count).sum())
    majority_prediction = 'tie' if n_at_max > 1 else int(counts.argmax())
    seed_agreement_rate = max_count / K

    if max_count == K:
        category = 'Highly Stable'
    elif max_count == K - 1:
        category = 'Stable'
    elif max_count == K - 2:
        category = 'Moderately Unstable'
    else:
        category = 'Highly Unstable / No Strong Consensus'

    return pd.Series({
        'majority_prediction': majority_prediction,
        'agreement_count': max_count,
        'seed_agreement_rate': seed_agreement_rate,
        'stability_category': category,
    })


def compute_prediction_entropy(row, pred_cols, class_names, K):
    """Section 7: Shannon entropy dari distribusi hard-prediction antar
    seed (pelengkap continuous untuk SAR/agreement_count yang diskrit).
    Term dengan p=0 diabaikan (0*ln(0) := 0)."""
    preds = row[pred_cols].values.astype(int)
    counts = np.bincount(preds, minlength=len(class_names))
    p = counts / K
    p_nonzero = p[p > 0]
    entropy = float(-np.sum(p_nonzero * np.log(p_nonzero)))
    max_entropy = math.log(len(class_names))
    normalized_entropy = entropy / max_entropy if max_entropy > 0 else 0.0
    return pd.Series({'prediction_entropy': entropy, 'normalized_entropy': normalized_entropy})


def compute_true_class_probability_stats(row, seeds, class_names):
    """Section 8: mean/SD probability yang diberikan model pada kelas TRUE
    gambar tsb, across 5 seed (confidence/probability stability, berbeda
    dari label-prediction stability)."""
    true_idx = int(row['true_label'])
    true_cname = class_names[true_idx]
    vals = np.array([row[f'prob_{true_cname}_seed_{s}'] for s in seeds], dtype=float)
    return pd.Series({
        'mean_probability_of_true_class': vals.mean(),
        'sd_probability_of_true_class': vals.std(ddof=1) if len(vals) > 1 else 0.0,
    })


def build_prediction_stability_df(test_df, run_artifacts, seeds, class_names):
    """
    # ============================================================
    # MULTI-SEED STABILITY IMPROVEMENT
    # ============================================================
    Gabungkan prediksi kelima seed lewat explicit merge pada `image_path`
    (bukan asumsi posisi baris). Setiap langkah merge divalidasi (identifier
    unik, jumlah sample sama, set identifier sama, true_label konsisten)
    dan GAGAL DENGAN ERROR jika dilanggar."""
    assert test_df['image_path'].is_unique, (
        'test_df image_path must be unique to use it as a merge identifier.'
    )
    base_paths = set(test_df['image_path'])
    K = len(seeds)

    stability_df = test_df[['image_path', 'label']].rename(columns={'label': 'true_label'}).copy()

    for seed in seeds:
        per_image = run_artifacts[seed]['per_image']

        assert len(per_image) == len(test_df), (
            f'Seed {seed}: got {len(per_image)} test predictions, expected {len(test_df)}.'
        )
        assert per_image['image_path'].is_unique, (
            f'Seed {seed}: duplicate image_path in predictions.'
        )
        assert set(per_image['image_path']) == base_paths, (
            f'Seed {seed}: predicted image_path set does not match the fixed test set '
            f'(missing or unexpected images).'
        )

        check = per_image[['image_path', 'true_label']].merge(
            test_df[['image_path', 'label']], on='image_path', how='left'
        )
        assert (check['true_label'] == check['label']).all(), (
            f'Seed {seed}: true_label recorded during evaluation does not match '
            f'test_df label for at least one image.'
        )

        seed_cols = {'image_path': per_image['image_path'].values,
                     f'pred_seed_{seed}': per_image['pred_label'].values}
        for cname in class_names:
            seed_cols[f'prob_{cname}_seed_{seed}'] = per_image[f'prob_{cname}'].values
        seed_df = pd.DataFrame(seed_cols)

        n_before = len(stability_df)
        stability_df = stability_df.merge(seed_df, on='image_path', how='left')
        assert len(stability_df) == n_before, (
            f'Seed {seed}: merge changed row count ({n_before} -> {len(stability_df)}) '
            f'-- image_path is not behaving as a 1:1 key.'
        )
        assert stability_df[f'pred_seed_{seed}'].notna().all(), (
            f'Seed {seed}: missing prediction for one or more test images after merge.'
        )

    pred_cols = [f'pred_seed_{s}' for s in seeds]

    # Section 6: SAR / majority prediction / 4-level stability category
    agreement_extra = stability_df.apply(
        lambda r: compute_agreement_and_majority(r, pred_cols, class_names, K), axis=1
    )
    stability_df = pd.concat([stability_df, agreement_extra], axis=1)

    # Section 7: prediction entropy (continuous instability measure)
    entropy_extra = stability_df.apply(
        lambda r: compute_prediction_entropy(r, pred_cols, class_names, K), axis=1
    )
    stability_df = pd.concat([stability_df, entropy_extra], axis=1)

    # Section 8: probability-based stability -- per-class mean/SD across
    # seeds, plus mean/SD of the probability assigned to the image's own
    # true class.
    for cname in class_names:
        prob_cols = [f'prob_{cname}_seed_{s}' for s in seeds]
        stability_df[f'mean_prob_{cname}'] = stability_df[prob_cols].mean(axis=1)
        stability_df[f'sd_prob_{cname}']   = stability_df[prob_cols].std(axis=1, ddof=1)

    true_class_extra = stability_df.apply(
        lambda r: compute_true_class_probability_stats(r, seeds, class_names), axis=1
    )
    stability_df = pd.concat([stability_df, true_class_extra], axis=1)

    # Section 9: correctness stability
    for seed in seeds:
        stability_df[f'correct_seed_{seed}'] = (
            stability_df[f'pred_seed_{seed}'] == stability_df['true_label']
        ).astype(int)
    correct_cols = [f'correct_seed_{s}' for s in seeds]
    stability_df['n_correct_seeds'] = stability_df[correct_cols].sum(axis=1)

    def _correctness_status(n):
        if n == K:
            return 'Correct in All Seeds'
        elif n == 0:
            return 'Wrong in All Seeds'
        return 'Seed-Sensitive Correctness'

    stability_df['correctness_status'] = stability_df['n_correct_seeds'].apply(_correctness_status)

    return stability_df


def calculate_classwise_stability_summary(stability_df, class_names):
    """classwise_stability_summary.csv: profil agreement/entropy label
    prediction, overall dan per true class."""
    groups = {'Overall': stability_df}
    for i, cname in enumerate(class_names):
        groups[cname] = stability_df[stability_df['true_label'] == i]

    categories = ['Highly Stable', 'Stable', 'Moderately Unstable',
                  'Highly Unstable / No Strong Consensus']
    cat_keys = ['highly_stable', 'stable', 'moderately_unstable', 'highly_unstable']

    rows = []
    for group_name, g in groups.items():
        n = len(g)
        if n == 0:
            continue
        row = {'group': group_name, 'n_images': n}
        for cat, key in zip(categories, cat_keys):
            cnt = int((g['stability_category'] == cat).sum())
            row[f'n_{key}'] = cnt
            row[f'pct_{key}'] = 100 * cnt / n
        n_tie = int((g['majority_prediction'] == 'tie').sum())
        row['n_tie_majority'] = n_tie
        row['pct_tie_majority'] = 100 * n_tie / n
        row['mean_seed_agreement_rate'] = g['seed_agreement_rate'].mean()
        row['mean_prediction_entropy'] = g['prediction_entropy'].mean()
        row['mean_normalized_entropy'] = g['normalized_entropy'].mean()
        rows.append(row)
    return pd.DataFrame(rows)


def calculate_prediction_stability_summary(stability_df, class_names):
    """prediction_stability_summary.csv: correctness stability (Section 9)
    dan probability/confidence stability (Section 8), overall dan per true
    class -- berbeda dari profil label-agreement di atas."""
    groups = {'Overall': stability_df}
    for i, cname in enumerate(class_names):
        groups[cname] = stability_df[stability_df['true_label'] == i]

    rows = []
    for group_name, g in groups.items():
        n = len(g)
        if n == 0:
            continue
        n_correct_all = int((g['correctness_status'] == 'Correct in All Seeds').sum())
        n_wrong_all   = int((g['correctness_status'] == 'Wrong in All Seeds').sum())
        n_sensitive   = int((g['correctness_status'] == 'Seed-Sensitive Correctness').sum())
        rows.append({
            'group': group_name, 'n_images': n,
            'n_correct_all_seeds': n_correct_all, 'pct_correct_all_seeds': 100 * n_correct_all / n,
            'n_wrong_all_seeds': n_wrong_all, 'pct_wrong_all_seeds': 100 * n_wrong_all / n,
            'n_seed_sensitive_correctness': n_sensitive,
            'pct_seed_sensitive_correctness': 100 * n_sensitive / n,
            'mean_n_correct_seeds': g['n_correct_seeds'].mean(),
            'mean_probability_of_true_class': g['mean_probability_of_true_class'].mean(),
            'mean_sd_probability_of_true_class': g['sd_probability_of_true_class'].mean(),
        })
    return pd.DataFrame(rows)


stability_df = build_prediction_stability_df(test_df, run_artifacts, SEEDS, CLASS_NAMES)
classwise_stability_df = calculate_classwise_stability_summary(stability_df, CLASS_NAMES)
prediction_stability_summary_df = calculate_prediction_stability_summary(stability_df, CLASS_NAMES)

print('\n' + '=' * 70)
print(' PREDICTION LABEL-AGREEMENT STABILITY (Overall & Per True Class) ')
print('=' * 70)
print(classwise_stability_df.to_string(index=False))

print('\n' + '=' * 70)
print(' CORRECTNESS & PROBABILITY STABILITY (Overall & Per True Class) ')
print('=' * 70)
print(prediction_stability_summary_df.to_string(index=False))

"""## 13. Save Result Files"""

def save_results(overall_df, classwise_df, summary_df, stability_df,
                  classwise_stability_df, prediction_stability_summary_df, output_dir):
    overall_df.to_csv(os.path.join(output_dir, 'multiseed_overall_results.csv'), index=False)
    classwise_df.to_csv(os.path.join(output_dir, 'multiseed_classwise_results.csv'), index=False)
    summary_df.to_csv(os.path.join(output_dir, 'multiseed_summary.csv'), index=False)
    stability_df.to_csv(os.path.join(output_dir, 'prediction_stability.csv'), index=False)
    classwise_stability_df.to_csv(
        os.path.join(output_dir, 'classwise_stability_summary.csv'), index=False
    )
    prediction_stability_summary_df.to_csv(
        os.path.join(output_dir, 'prediction_stability_summary.csv'), index=False
    )
    print(f'Semua file CSV hasil eksperimen tersimpan di: {output_dir}')
    print(' - multiseed_overall_results.csv')
    print(' - multiseed_classwise_results.csv')
    print(' - multiseed_summary.csv')
    print(' - prediction_stability.csv')
    print(' - classwise_stability_summary.csv')
    print(' - prediction_stability_summary.csv')
    print(' - run_manifest.csv (tersimpan di Section 10)')
    print(' - reproducibility_check.csv (tersimpan di Section 9b, bila RUN_REPEATABILITY_CHECK=True)')
    print(' - experiment_environment.txt (tersimpan di Section 9 & akhir eksperimen)')
    print(' - training_history_seed{42,123,2026,3407,8901}.csv (tersimpan per-seed saat training)')


save_results(overall_df, classwise_df, summary_df, stability_df,
             classwise_stability_df, prediction_stability_summary_df, OUTPUT_DIR)


def save_and_report_computational_benchmark(
    scenario_id, scenario_label, dataprep_engine, dataprep_duration_sec, dataprep_total_images,
    overall_df, output_dir, drive_dir=None, t_pipeline_start=None
):
    """Menghitung dan mengekspor metrik komputasi, waktu per tahapan,
    throughput, dan utilisasi resource (VRAM) per skenario."""
    total_pipeline_time_sec = (time.time() - t_pipeline_start) if t_pipeline_start else 0.0
    
    dataprep_throughput = dataprep_total_images / dataprep_duration_sec if dataprep_duration_sec > 0 else 0.0
    train_times = overall_df['training_time_sec'].values
    mean_train_time = float(np.mean(train_times))
    std_train_time = float(np.std(train_times, ddof=1)) if len(train_times) > 1 else 0.0
    mean_epoch_time = mean_train_time / MAX_EPOCHS
    
    inf_times = overall_df['inference_time_sec'].values if 'inference_time_sec' in overall_df.columns else [0.0]
    mean_inf_time = float(np.mean(inf_times))
    std_inf_time = float(np.std(inf_times, ddof=1)) if len(inf_times) > 1 else 0.0
    test_samples = len(test_df)
    inf_throughput = test_samples / mean_inf_time if mean_inf_time > 0 else 0.0
    
    peak_vram = float(overall_df['peak_gpu_mem_mb'].max()) if 'peak_gpu_mem_mb' in overall_df.columns else 0.0
    
    benchmark_data = [
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '1_data_management',
         'task': 'leakage_and_deduplication', 'engine': dataprep_engine, 'metric': 'duration_seconds', 'value': round(dataprep_duration_sec, 2), 'unit': 'seconds'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '1_data_management',
         'task': 'leakage_and_deduplication', 'engine': dataprep_engine, 'metric': 'throughput', 'value': round(dataprep_throughput, 1), 'unit': 'images/sec'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '2_model_training',
         'task': '5seeds_training', 'engine': 'PyTorch GPU (Tesla T4)', 'metric': 'mean_seed_duration_seconds', 'value': round(mean_train_time, 2), 'unit': 'seconds'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '2_model_training',
         'task': '5seeds_training', 'engine': 'PyTorch GPU (Tesla T4)', 'metric': 'std_seed_duration_seconds', 'value': round(std_train_time, 2), 'unit': 'seconds'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '2_model_training',
         'task': 'epoch_training', 'engine': 'PyTorch GPU (Tesla T4)', 'metric': 'mean_epoch_seconds', 'value': round(mean_epoch_time, 2), 'unit': 'seconds'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '2_model_training',
         'task': 'gpu_resource', 'engine': 'CUDA', 'metric': 'peak_gpu_vram_mb', 'value': round(peak_vram, 1), 'unit': 'MB'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '3_test_inference',
         'task': 'batch_inference_and_eval', 'engine': INFERENCE_ENGINE_LABEL, 'metric': 'mean_inference_seconds', 'value': round(mean_inf_time, 2), 'unit': 'seconds'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '3_test_inference',
         'task': 'batch_inference_and_eval', 'engine': INFERENCE_ENGINE_LABEL, 'metric': 'inference_throughput', 'value': round(inf_throughput, 1), 'unit': 'images/sec'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '4_pipeline_total',
         'task': 'end_to_end_wallclock', 'engine': 'Pipeline Total', 'metric': 'total_wallclock_seconds', 'value': round(total_pipeline_time_sec, 2), 'unit': 'seconds'},
        {'scenario_id': scenario_id, 'scenario_name': scenario_label, 'stage': '4_pipeline_total',
         'task': 'end_to_end_wallclock', 'engine': 'Pipeline Total', 'metric': 'total_wallclock_minutes', 'value': round(total_pipeline_time_sec / 60.0, 2), 'unit': 'minutes'},
    ]
    
    bench_df = pd.DataFrame(benchmark_data)
    bench_csv = os.path.join(output_dir, f'computational_benchmark_{scenario_id}.csv')
    bench_df.to_csv(bench_csv, index=False)
    
    # Visualisasi Profil Komputasi
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    stages = ['Data Mgmt\n(8,323 imgs)', 'Avg Training / Seed\n(6,973 imgs x 10 ep)', 'Avg Test Inference\n(600 imgs)']
    times = [dataprep_duration_sec, mean_train_time, mean_inf_time]
    colors = ['#3498db', '#e74c3c', '#2ecc71']
    
    bars = axes[0].bar(stages, times, color=colors, edgecolor='black', linewidth=1.2)
    axes[0].set_ylabel('Waktu Eksekusi (Detik)', fontsize=11, fontweight='bold')
    axes[0].set_title(f'Waktu Eksekusi Per Tahapan ({scenario_id})', fontsize=12, fontweight='bold')
    axes[0].grid(axis='y', linestyle='--', alpha=0.5)
    for bar in bars:
        h = bar.get_height()
        axes[0].text(bar.get_x() + bar.get_width()/2., h + (max(times)*0.01 if max(times)>0 else 0.1), f'{h:.1f}s', ha='center', va='bottom', fontweight='bold')

    throughputs = [dataprep_throughput, (6973 * 10) / mean_train_time if mean_train_time>0 else 0, inf_throughput]
    t_labels = ['Data Mgmt\n(imgs/s)', 'Training\n(imgs/s)', 'Inference\n(imgs/s)']
    bars2 = axes[1].bar(t_labels, throughputs, color=['#9b59b6', '#f39c12', '#1abc9c'], edgecolor='black', linewidth=1.2)
    axes[1].set_ylabel('Throughput (Samples / Detik)', fontsize=11, fontweight='bold')
    axes[1].set_title(f'Throughput Komputasi ({scenario_id})', fontsize=12, fontweight='bold')
    axes[1].grid(axis='y', linestyle='--', alpha=0.5)
    for bar in bars2:
        h = bar.get_height()
        axes[1].text(bar.get_x() + bar.get_width()/2., h + (max(throughputs)*0.01 if max(throughputs)>0 else 0.1), f'{h:.1f}', ha='center', va='bottom', fontweight='bold')
        
    plt.tight_layout()
    plot_path = os.path.join(output_dir, f'fig_computational_profile_{scenario_id}.png')
    plt.savefig(plot_path, dpi=150, bbox_inches='tight')
    plt.close()

    # Executive Summary Box di Console
    print('\n' + '=' * 80)
    print(f'🏆 RINGKASAN BENCHMARK KOMPUTASI: {scenario_label.upper()}')
    print('=' * 80)
    print(f'1. Data Management (8,323 images):')
    print(f'   - Engine              : {dataprep_engine}')
    print(f'   - Waktu Eksekusi      : {dataprep_duration_sec:.2f} detik')
    print(f'   - Throughput          : {dataprep_throughput:.1f} images/detik')
    print(f'2. Model Training (5 Seeds x 10 Epochs):')
    print(f'   - Engine              : PyTorch GPU (Tesla T4)')
    print(f'   - Rata-rata per Seed  : {mean_train_time:.2f} detik (± {std_train_time:.2f})')
    print(f'   - Rata-rata per Epoch : {mean_epoch_time:.2f} detik')
    print(f'   - Peak GPU Memory     : {peak_vram:.1f} MB')
    print(f'3. Test Inference & Evaluasi (600 test images):')
    print(f'   - Engine              : {INFERENCE_ENGINE_LABEL}')
    print(f'   - Rata-rata per Seed  : {mean_inf_time:.2f} detik')
    print(f'   - Throughput          : {inf_throughput:.1f} images/detik')
    print(f'4. Total Pipeline Wall-Clock Time:')
    print(f'   - Durasi Total        : {total_pipeline_time_sec/60.0:.2f} menit ({total_pipeline_time_sec:.1f} detik)')
    print('=' * 80)

    # Sinkronisasi / Backup ke Google Drive
    if drive_dir and os.path.exists('/content/drive/MyDrive'):
        import shutil
        try:
            shutil.copytree(output_dir, drive_dir, dirs_exist_ok=True)
            print(f'>>> Semua output berhasil disinkronkan ke Google Drive: {drive_dir} <<<')
        except Exception as e:
            print(f'Peringatan sinkronisasi drive: {e}')

    # Cek apakah ada file benchmark dari skenario lain di Google Drive
    try:
        drive_root = '/content/drive/MyDrive/skin_lesion_project'
        if os.path.exists(drive_root):
            bench_files = []
            for root, _, files in os.walk(drive_root):
                for f in files:
                    if f.startswith('computational_benchmark_') and f.endswith('.csv'):
                        bench_files.append(os.path.join(root, f))
            if len(bench_files) >= 2:
                dfs = [pd.read_csv(bf) for bf in bench_files]
                combined_bench = pd.concat(dfs, ignore_index=True).drop_duplicates()
                comb_path = os.path.join(drive_root, 'COMPARATIVE_BENCHMARK_ALL_SCENARIOS.csv')
                combined_bench.to_csv(comb_path, index=False)
                print(f'\n[Multi-Scenario Detected] File perbandingan lintas-skenario terupdate: {comb_path}')
    except Exception:
        pass


save_and_report_computational_benchmark(
    SCENARIO_ID, SCENARIO_LABEL, DATAPREP_ENGINE_LABEL, dataprep_duration_sec,
    len(all_df), overall_df, OUTPUT_DIR, DRIVE_OUTPUT_DIR, t_pipeline_start
)

"""## 14. Visualization

Visualisasi untuk publikasi ilmiah (tidak ada grafik 3D): boxplot distribusi
metrik, error-bar class-wise F1/Recall, distribusi Seed Agreement Rate,
heatmap agreement antar-seed, bar chart kategori stabilitas 4-level,
distribusi entropy per true class, distribusi n_correct_seeds per true
class, dan training/validation curves per seed.
"""

plt.rcParams.update({
    'font.size': 11, 'axes.titlesize': 12, 'axes.labelsize': 11,
    'legend.fontsize': 9, 'figure.dpi': 100,
})


def plot_metric_distribution(overall_df, output_dir):
    fig, axes = plt.subplots(1, 3, figsize=(13, 4.5))
    metrics = [('accuracy', 'Accuracy'), ('macro_f1', 'Macro F1-score'), ('macro_auc', 'Macro ROC-AUC')]
    for ax, (col, label) in zip(axes, metrics):
        data = overall_df[col].values
        ax.boxplot(data, widths=0.5, showmeans=True)
        jitter = np.random.default_rng(0).normal(1, 0.02, size=len(data))
        ax.scatter(jitter, data, color='#1f77b4', alpha=0.85, zorder=3, s=30)
        ax.set_title(label)
        ax.set_xticks([1]); ax.set_xticklabels(['EfficientNet-B0'])
        ax.set_ylabel(label)
        ax.grid(axis='y', alpha=0.3)
    fig.suptitle('Distribution of Test Performance Across 5 Controlled Random Seeds')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'fig_metric_distribution_boxplot.png'),
                dpi=150, bbox_inches='tight')
    plt.show()


def plot_classwise_errorbar(summary_df, metric_name, class_names, output_dir, ylabel):
    sub = summary_df[(summary_df['scope'] == 'classwise') & (summary_df['metric'] == metric_name)]
    sub = sub.set_index('class_name').loc[class_names].reset_index()
    x = np.arange(len(class_names))
    fig, ax = plt.subplots(figsize=(6.5, 4.5))
    ax.errorbar(x, sub['mean'], yerr=sub['sd'], fmt='o', capsize=5, markersize=7,
                color='#1f77b4', ecolor='#1f77b4', elinewidth=1.5)
    ax.set_xticks(x)
    ax.set_xticklabels([c.replace('_', ' ').title() for c in class_names])
    ax.set_ylabel(ylabel)
    ax.set_title(f'Class-wise {ylabel} (Mean ± SD across 5 seeds)')
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    fname = f'fig_classwise_{metric_name}_errorbar.png'
    plt.savefig(os.path.join(output_dir, fname), dpi=150, bbox_inches='tight')
    plt.show()


def plot_sar_distribution(stability_df, output_dir):
    """Section 15.D: Seed Agreement Rate distribution."""
    counts = stability_df['seed_agreement_rate'].value_counts().sort_index()
    fig, ax = plt.subplots(figsize=(7, 4.5))
    ax.bar([f'{v:.1f}' for v in counts.index], counts.values, color='#1f77b4', width=0.6)
    for i, v in enumerate(counts.values):
        ax.text(i, v, str(int(v)), ha='center', va='bottom', fontsize=9)
    ax.set_xlabel('Seed Agreement Rate (agreement_count / 5)')
    ax.set_ylabel('Number of test images')
    ax.set_title('Distribution of Seed Agreement Rate Across Test Images')
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'fig_sar_distribution.png'), dpi=150, bbox_inches='tight')
    plt.show()


def plot_seed_agreement_heatmap(stability_df, seeds, output_dir):
    """Supplementary: pairwise prediction agreement between seed pairs."""
    K = len(seeds)
    agreement = np.zeros((K, K))
    for i, si in enumerate(seeds):
        for j, sj in enumerate(seeds):
            agreement[i, j] = (stability_df[f'pred_seed_{si}'] == stability_df[f'pred_seed_{sj}']).mean()
    fig, ax = plt.subplots(figsize=(6, 5))
    sns.heatmap(agreement, annot=True, fmt='.3f', cmap='Blues',
                xticklabels=[str(s) for s in seeds], yticklabels=[str(s) for s in seeds],
                vmin=0, vmax=1, ax=ax, cbar_kws={'label': 'Prediction agreement rate'})
    ax.set_xlabel('Seed'); ax.set_ylabel('Seed')
    ax.set_title('Pairwise Test-Set Prediction Agreement Between Seeds')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'fig_seed_agreement_heatmap.png'),
                dpi=150, bbox_inches='tight')
    plt.show()


def plot_stability_categories_bar(classwise_stability_df, output_dir):
    """Section 15.E: 4-level stability category counts, overall & per class."""
    categories = ['n_highly_stable', 'n_stable', 'n_moderately_unstable', 'n_highly_unstable']
    cat_labels = ['Highly Stable (5/5)', 'Stable (4/5)', 'Moderately Unstable (3/5)',
                  'Highly Unstable / No Consensus (<=2/5)']
    groups = classwise_stability_df['group'].tolist()
    x = np.arange(len(groups))
    width = 0.19
    offsets = np.linspace(-1.5, 1.5, 4) * width
    colors = ['#2ca02c', '#8bc34a', '#ff9800', '#d62728']
    fig, ax = plt.subplots(figsize=(9.5, 5))
    for off, col, lab, color in zip(offsets, categories, cat_labels, colors):
        ax.bar(x + off, classwise_stability_df[col], width, label=lab, color=color)
    ax.set_xticks(x); ax.set_xticklabels(groups, rotation=15, ha='right')
    ax.set_ylabel('Number of test images')
    ax.set_title('Prediction Stability Categories — Overall and Per True Class')
    ax.legend(fontsize=8)
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'fig_stability_categories_bar.png'),
                dpi=150, bbox_inches='tight')
    plt.show()


def plot_entropy_distribution_by_class(stability_df, class_names, output_dir):
    """Section 15.F: prediction entropy distribution per true class."""
    fig, ax = plt.subplots(figsize=(7.5, 4.5))
    data = [stability_df.loc[stability_df['true_label'] == i, 'prediction_entropy'].values
            for i in range(len(class_names))]
    ax.boxplot(data, labels=[c.replace('_', ' ').title() for c in class_names], showmeans=True)
    ax.set_ylabel('Prediction entropy (nats)')
    ax.set_title('Prediction Entropy Across Seeds, by True Class')
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'fig_entropy_by_class.png'), dpi=150, bbox_inches='tight')
    plt.show()


def plot_ncorrect_distribution_by_class(stability_df, class_names, output_dir):
    """Section 15.G: n_correct_seeds distribution per true class."""
    K_values = sorted(stability_df['n_correct_seeds'].unique())
    x = np.arange(len(K_values))
    width = 0.8 / len(class_names)
    fig, ax = plt.subplots(figsize=(8.5, 5))
    for ci, cname in enumerate(class_names):
        sub = stability_df[stability_df['true_label'] == ci]
        counts = [(sub['n_correct_seeds'] == k).sum() for k in K_values]
        ax.bar(x + (ci - (len(class_names) - 1) / 2) * width, counts, width,
               label=cname.replace('_', ' ').title())
    ax.set_xticks(x); ax.set_xticklabels([str(k) for k in K_values])
    ax.set_xlabel('Number of seeds with a correct prediction (n_correct_seeds)')
    ax.set_ylabel('Number of test images')
    ax.set_title('Correctness Stability Across Seeds, by True Class')
    ax.legend()
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'fig_ncorrect_by_class.png'), dpi=150, bbox_inches='tight')
    plt.show()


def plot_training_curves(seeds, output_dir):
    fig, axes = plt.subplots(len(seeds), 2, figsize=(11, 3.0 * len(seeds)))
    for row, seed in enumerate(seeds):
        hist = pd.read_csv(os.path.join(output_dir, f'training_history_seed{seed}.csv'))
        ax_loss, ax_auc = axes[row, 0], axes[row, 1]
        ax_loss.plot(hist['epoch'], hist['train_loss'], label='Train Loss', color='#1f77b4')
        ax_loss.plot(hist['epoch'], hist['val_loss'], label='Val Loss', color='#d62728')
        ax_loss.set_ylabel(f'Seed {seed}\nLoss'); ax_loss.grid(alpha=0.3)
        if row == 0:
            ax_loss.legend()
        ax_auc.plot(hist['epoch'], hist['val_auc'], label='Val Macro-AUC', color='#2ca02c')
        ax_auc.set_ylabel('Macro-AUC'); ax_auc.grid(alpha=0.3)
        if row == 0:
            ax_auc.legend()
        if row == len(seeds) - 1:
            ax_loss.set_xlabel('Epoch'); ax_auc.set_xlabel('Epoch')
    fig.suptitle('Training / Validation Curves per Seed')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'fig_training_curves_per_seed.png'),
                dpi=150, bbox_inches='tight')
    plt.show()


def plot_stability_results(overall_df, summary_df, stability_df,
                            classwise_stability_df, seeds, class_names, output_dir):
    plot_metric_distribution(overall_df, output_dir)
    plot_classwise_errorbar(summary_df, 'f1', class_names, output_dir, 'F1-score')
    plot_classwise_errorbar(summary_df, 'recall', class_names, output_dir, 'Recall')
    plot_sar_distribution(stability_df, output_dir)
    plot_seed_agreement_heatmap(stability_df, seeds, output_dir)
    plot_stability_categories_bar(classwise_stability_df, output_dir)
    plot_entropy_distribution_by_class(stability_df, class_names, output_dir)
    plot_ncorrect_distribution_by_class(stability_df, class_names, output_dir)
    plot_training_curves(seeds, output_dir)


plot_stability_results(overall_df, summary_df, stability_df, classwise_stability_df,
                        SEEDS, CLASS_NAMES, OUTPUT_DIR)

save_environment_info('END')

"""## 15. Supplementary: Explainable AI (XAI) Visualization (Illustrative Only)

PENTING: bagian ini BUKAN bagian dari hasil utama multi-seed stability
analysis di atas, dan TIDAK memengaruhi metrik/statistik yang sudah
dihitung ataupun model selection. `XAI_ILLUSTRATIVE_SEED` adalah seed
PREDEFINED 42 (bukan dipilih berdasarkan performa test) -- lihat Section 16
review spec. Sample gambar test yang dibaca di bawah HANYA dipakai untuk
visualisasi post-hoc (Section 17 review spec): tidak memengaruhi training,
early stopping, atau model selection manapun, walau ini berarti test set
DIBACA LAGI setelah evaluasi multi-seed di atas (jadi tidak benar untuk
mengklaim test set "hanya pernah diakses sekali").
"""

!pip install -q grad-cam captum lime shap

import torch.nn.functional as Fnn
from pytorch_grad_cam import GradCAM, GradCAMPlusPlus, EigenCAM, GuidedBackpropReLUModel
from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget
from pytorch_grad_cam.utils.image import show_cam_on_image, deprocess_image
from captum.attr import Saliency, IntegratedGradients
import shap
from lime import lime_image
from skimage.segmentation import mark_boundaries

print('Library XAI siap dipakai.')

# Predefined fixed seed (NOT selected by test performance) -- Section 16.
XAI_ILLUSTRATIVE_SEED = 42

xai_model = EfficientNetB0Classifier(NUM_CLASSES, pretrained=False).to(DEVICE)
xai_ckpt_path = os.path.join(OUTPUT_DIR, f'efficientnet_b0_seed{XAI_ILLUSTRATIVE_SEED}_best.pth')
xai_model.load_state_dict(torch.load(xai_ckpt_path, map_location=DEVICE))
xai_model.eval()
print(f'XAI illustrative model loaded from: {xai_ckpt_path}')


def denormalize(tensor_img):
    """tensor_img: torch.Tensor shape (3, H, W), hasil ToTensor+Normalize.
    Mengembalikan numpy array (H, W, 3) float [0,1] siap ditampilkan."""
    mean = np.array(IMAGENET_MEAN).reshape(3, 1, 1)
    std  = np.array(IMAGENET_STD).reshape(3, 1, 1)
    img = tensor_img.cpu().numpy() * std + mean
    img = np.clip(img, 0, 1)
    return np.transpose(img, (1, 2, 0))


def get_sample_per_class(df, n_per_class=1, seed=DATA_PREP_SEED):
    """Ambil n_per_class sample per kelas dari dataframe (test_df), dengan
    seed TETAP (DATA_PREP_SEED) -- tidak terikat ke seed training manapun.
    Post-hoc visualization only: dipanggil SETELAH seluruh evaluasi/model
    selection multi-seed selesai, dan tidak memengaruhinya sama sekali."""
    samples = []
    for label_val in sorted(df['label'].unique()):
        subset = df[df['label'] == label_val].sample(n=n_per_class, random_state=seed)
        samples.append(subset)
    return pd.concat(samples, ignore_index=True)


xai_samples_df = get_sample_per_class(test_df, n_per_class=1)
print(f'Jumlah sample untuk XAI: {len(xai_samples_df)}')
print(xai_samples_df[['image_path', 'dx', 'label']])

xai_dataset = HAMDataset(xai_samples_df, transform=val_test_transform)

target_layers = [xai_model.backbone.conv_head]
cam_methods = {
    'Grad-CAM':   GradCAM(model=xai_model, target_layers=target_layers),
    'Grad-CAM++': GradCAMPlusPlus(model=xai_model, target_layers=target_layers),
    'Eigen-CAM':  EigenCAM(model=xai_model, target_layers=target_layers),
}
guided_bp = GuidedBackpropReLUModel(model=xai_model, device=DEVICE)


def visualize_cam_methods(idx):
    img_tensor, true_label = xai_dataset[idx]
    input_tensor = img_tensor.unsqueeze(0).to(DEVICE)
    rgb_img = denormalize(img_tensor)

    with torch.no_grad():
        logits = xai_model(input_tensor)
        pred_label = logits.argmax(dim=1).item()

    targets = [ClassifierOutputTarget(pred_label)]

    fig, axes = plt.subplots(1, 5, figsize=(20, 4.5))
    axes[0].imshow(rgb_img); axes[0].set_title('Original'); axes[0].axis('off')

    col = 1
    cam_masks = {}
    for name, cam in cam_methods.items():
        grayscale_cam = cam(input_tensor=input_tensor, targets=targets)[0]
        cam_masks[name] = grayscale_cam
        vis = show_cam_on_image(rgb_img, grayscale_cam, use_rgb=True)
        axes[col].imshow(vis); axes[col].set_title(name); axes[col].axis('off')
        col += 1

    gb = guided_bp(input_tensor, target_category=pred_label)
    gb = gb - gb.min(); gb = gb / (gb.max() + 1e-8)
    cam_mask_resized = cam_masks['Grad-CAM'][..., None]
    guided_gradcam = deprocess_image(gb * cam_mask_resized)
    axes[col].imshow(guided_gradcam); axes[col].set_title('Guided Grad-CAM'); axes[col].axis('off')

    fig.suptitle(f'[Seed {XAI_ILLUSTRATIVE_SEED}, illustrative] '
                  f'True: {CLASS_NAMES[true_label]} | Pred: {CLASS_NAMES[pred_label]}', fontsize=12)
    plt.tight_layout()
    plt.savefig(f'{OUTPUT_DIR}/xai_cam_methods_sample{idx}.png', dpi=130, bbox_inches='tight')
    plt.show()


for i in range(len(xai_dataset)):
    visualize_cam_methods(i)

saliency_method = Saliency(xai_model)
ig_method = IntegratedGradients(xai_model)


def visualize_gradient_methods(idx):
    img_tensor, true_label = xai_dataset[idx]
    input_tensor = img_tensor.unsqueeze(0).to(DEVICE)
    input_tensor.requires_grad_()
    rgb_img = denormalize(img_tensor)

    with torch.no_grad():
        pred_label = xai_model(input_tensor).argmax(dim=1).item()

    grads = saliency_method.attribute(input_tensor, target=pred_label)
    saliency_map = grads.squeeze().cpu().detach().numpy()
    saliency_map = np.abs(saliency_map).max(axis=0)
    saliency_map = (saliency_map - saliency_map.min()) / (saliency_map.max() - saliency_map.min() + 1e-8)

    baseline = torch.zeros_like(input_tensor)
    attributions_ig = ig_method.attribute(input_tensor, baseline, target=pred_label, n_steps=50)
    ig_map = attributions_ig.squeeze().cpu().detach().numpy()
    ig_map = np.abs(ig_map).sum(axis=0)
    ig_map = (ig_map - ig_map.min()) / (ig_map.max() - ig_map.min() + 1e-8)

    fig, axes = plt.subplots(1, 3, figsize=(13, 4.5))
    axes[0].imshow(rgb_img); axes[0].set_title('Original'); axes[0].axis('off')
    axes[1].imshow(saliency_map, cmap='hot'); axes[1].set_title('Saliency Map'); axes[1].axis('off')
    axes[2].imshow(ig_map, cmap='hot'); axes[2].set_title('Integrated Gradients'); axes[2].axis('off')

    fig.suptitle(f'[Seed {XAI_ILLUSTRATIVE_SEED}, illustrative] '
                  f'True: {CLASS_NAMES[true_label]} | Pred: {CLASS_NAMES[pred_label]}', fontsize=12)
    plt.tight_layout()
    plt.savefig(f'{OUTPUT_DIR}/xai_gradient_methods_sample{idx}.png', dpi=130, bbox_inches='tight')
    plt.show()


for i in range(len(xai_dataset)):
    visualize_gradient_methods(i)

# SHAP background: sample kecil DETERMINISTIK dari train set (fixed
# DATA_PREP_SEED, transform tanpa augmentasi) khusus untuk baseline SHAP.
xai_background_loader = DataLoader(
    HAMDataset(train_df.sample(n=min(64, len(train_df)), random_state=DATA_PREP_SEED),
               transform=val_test_transform),
    batch_size=20, shuffle=False,
)
background_batch = next(iter(xai_background_loader))[0].to(DEVICE)
shap_explainer = shap.GradientExplainer(xai_model, background_batch)


def visualize_shap(idx):
    img_tensor, true_label = xai_dataset[idx]
    input_tensor = img_tensor.unsqueeze(0).to(DEVICE)
    rgb_img = denormalize(img_tensor)

    with torch.no_grad():
        pred_label = xai_model(input_tensor).argmax(dim=1).item()

    shap_values = shap_explainer.shap_values(input_tensor, nsamples=50)

    if isinstance(shap_values, list):
        sv_pred = shap_values[pred_label][0]
    else:
        sv_pred = shap_values[0, ..., pred_label]

    sv_map = np.abs(sv_pred).sum(axis=0)
    sv_map = (sv_map - sv_map.min()) / (sv_map.max() - sv_map.min() + 1e-8)

    fig, axes = plt.subplots(1, 2, figsize=(9, 4.5))
    axes[0].imshow(rgb_img); axes[0].set_title('Original'); axes[0].axis('off')
    axes[1].imshow(sv_map, cmap='bwr'); axes[1].set_title('SHAP'); axes[1].axis('off')
    fig.suptitle(f'[Seed {XAI_ILLUSTRATIVE_SEED}, illustrative] '
                  f'True: {CLASS_NAMES[true_label]} | Pred: {CLASS_NAMES[pred_label]}', fontsize=12)
    plt.tight_layout()
    plt.savefig(f'{OUTPUT_DIR}/xai_shap_sample{idx}.png', dpi=130, bbox_inches='tight')
    plt.show()


for i in range(len(xai_dataset)):
    visualize_shap(i)

lime_explainer = lime_image.LimeImageExplainer()


def batch_predict_for_lime(images_np):
    """images_np: array (N, H, W, 3) uint8 [0,255] dari LIME."""
    mean = np.array(IMAGENET_MEAN)
    std  = np.array(IMAGENET_STD)
    batch = []
    for img in images_np:
        img = img.astype(np.float32) / 255.0
        img = (img - mean) / std
        img = np.transpose(img, (2, 0, 1))
        batch.append(img)
    batch_tensor = torch.tensor(np.stack(batch), dtype=torch.float32).to(DEVICE)
    with torch.no_grad():
        logits = xai_model(batch_tensor)
        probs = torch.softmax(logits, dim=1)
    return probs.cpu().numpy()


def visualize_lime(idx):
    img_tensor, true_label = xai_dataset[idx]
    rgb_img = denormalize(img_tensor)
    rgb_img_uint8 = (rgb_img * 255).astype(np.uint8)

    with torch.no_grad():
        pred_label = xai_model(img_tensor.unsqueeze(0).to(DEVICE)).argmax(dim=1).item()

    explanation = lime_explainer.explain_instance(
        rgb_img_uint8, batch_predict_for_lime,
        top_labels=NUM_CLASSES, hide_color=0, num_samples=500
    )
    lime_img, mask = explanation.get_image_and_mask(
        pred_label, positive_only=True, num_features=8, hide_rest=False
    )

    fig, axes = plt.subplots(1, 2, figsize=(9, 4.5))
    axes[0].imshow(rgb_img); axes[0].set_title('Original'); axes[0].axis('off')
    axes[1].imshow(mark_boundaries(lime_img, mask)); axes[1].set_title('LIME'); axes[1].axis('off')
    fig.suptitle(f'[Seed {XAI_ILLUSTRATIVE_SEED}, illustrative] '
                  f'True: {CLASS_NAMES[true_label]} | Pred: {CLASS_NAMES[pred_label]}', fontsize=12)
    plt.tight_layout()
    plt.savefig(f'{OUTPUT_DIR}/xai_lime_sample{idx}.png', dpi=130, bbox_inches='tight')
    plt.show()


for i in range(len(xai_dataset)):
    visualize_lime(i)

print('\nInterpretation reminder: controlled random seeds are used here to QUANTIFY')
print('stochastic performance variability and class-wise/image-level stability of')
print('EfficientNet-B0 -- NOT to search for a best seed, and NOT as evidence that')
print('multi-seed training improves accuracy.')
print('\nSelesai. Semua output (checkpoint, CSV, figure, environment info) tersimpan di:', OUTPUT_DIR)

# Kompresi folder output menjadi 1 file ZIP agar mudah diunduh di tab Output Kaggle
import shutil
zip_target = f'/kaggle/working/{SCENARIO_ID}_results'
try:
    shutil.make_archive(zip_target, 'zip', OUTPUT_DIR)
    print(f'\n📦 File ZIP hasil berhasil dibuat: {zip_target}.zip')
    print('👉 Anda bisa langsung klik tombol Tiga Titik (...) di samping file zip pada panel Output untuk download!')
except Exception as e:
    print(f'Peringatan kompresi ZIP: {e}')
```
